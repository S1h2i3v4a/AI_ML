# Lecture 19.4: Box Plots
**Topics Covered:**
- Five-Number Summary (Min, Q1, Median/Q2, Q3, Max)
- Interquartile Range (IQR = Q3 - Q1)
- Tukey's Outlier Detection Rule (1.5 * IQR)
- Basic `plt.boxplot()` usage

```python
import matplotlib.pyplot as plt
import numpy as np

# Lecture dataset
data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

plt.figure(figsize=(6, 6))
plt.boxplot(data)
plt.title('Basic Box Plot', fontsize=14, fontweight='bold')
plt.ylabel('Values', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Statistical calculation check
q1 = np.percentile(data, 25)
median = np.median(data)
q3 = np.percentile(data, 75)
iqr = q3 - q1
print(f'Q1 (25%): {q1}')
print(f'Median (50%): {median}')
print(f'Q3 (75%): {q3}')
print(f'IQR: {iqr}')
print(f'Lower Bound: {q1 - 1.5 * iqr}')
print(f'Upper Bound: {q3 + 1.5 * iqr}')
```
