# What is Gradient Descent in Linear Regression? (Formulas & Learning Rate)

Interactive notebook for Module 23.11: What is Gradient Descent in Linear Regression? (Formulas & Learning Rate).


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic linear data
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1) * 0.5
m = len(X)

# Hyperparameters
alpha = 0.1
n_iterations = 200

# Initialize parameters randomly
w = np.random.randn(1, 1)
b = np.random.randn(1, 1)

cost_history = []
w_history = [w[0, 0]]
b_history = [b[0, 0]]

# Gradient Descent Loop
for iteration in range(n_iterations):
    y_pred = X.dot(w) + b
    error = y_pred - y
    
    # Cost calculation
    cost = (1 / (2 * m)) * np.sum(error ** 2)
    cost_history.append(cost)
    
    # Gradients
    dj_dw = (1 / m) * X.T.dot(error)
    dj_db = (1 / m) * np.sum(error)
    
    # Simultaneous update
    w = w - alpha * dj_dw
    b = b - alpha * dj_db
    
    w_history.append(w[0, 0])
    b_history.append(b[0, 0])

print(f"Converged Weight (w): {w[0, 0]:.4f}")
print(f"Converged Bias (b): {b[0, 0]:.4f}")
print(f"Final Cost: {cost_history[-1]:.4f}")

# Plot Cost Convergence Curve
plt.figure(figsize=(8, 4.5))
plt.plot(range(n_iterations), cost_history, color="#2563eb", linewidth=2.5)
plt.title("Gradient Descent: Cost Function Convergence", fontsize=14, fontweight="bold")
plt.xlabel("Iteration", fontsize=12)
plt.ylabel("Cost J(w, b)", fontsize=12)
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
