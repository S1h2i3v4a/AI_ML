# Lecture 18.9: Bar Charts (Vertical)
**Topics Covered:**
- Concept of vertical bar charts
- `plt.bar(x, height)` syntax and parameters
- Comparing discrete categories with custom colors and widths

```python
import matplotlib.pyplot as plt

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]  # in $M

plt.figure(figsize=(8, 5))
plt.bar(years, revenue, color='purple', edgecolor='black', width=0.6)
plt.xlabel('Years', fontsize=12)
plt.ylabel('Revenue (in $M)', fontsize=12)
plt.title('Yearly Revenue of Movies', fontsize=14, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
