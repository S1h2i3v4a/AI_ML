# Understanding the Cost Function Curve (Convexity & Contour Plots)

Interactive notebook for Module 23.10: Understanding the Cost Function Curve (Convexity & Contour Plots).


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
np.random.seed(42)
x = 2 * np.random.rand(100)
y = 4 + 3 * x + np.random.randn(100) * 0.5
m = len(x)

# Create parameter grid
w_range = np.linspace(1, 5, 100)
b_range = np.linspace(2, 6, 100)
W, B = np.meshgrid(w_range, b_range)

# Vectorized cost computation
J = np.zeros(W.shape)
for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        y_pred = W[i, j] * x + B[i, j]
        J[i, j] = (1 / (2 * m)) * np.sum((y_pred - y) ** 2)

# Optimal parameters via Normal Equation
X_b = np.c_[np.ones((m, 1)), x]
theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
b_opt, w_opt = theta_best[0], theta_best[1]

# Plot Contour
plt.figure(figsize=(8, 6))
cs = plt.contour(W, B, J, levels=30, cmap="viridis")
plt.clabel(cs, inline=1, fontsize=9)
plt.scatter(w_opt, b_opt, color="#dc2626", s=140, marker="*", label=f"Global Min: w={w_opt:.2f}, b={b_opt:.2f}")

plt.title("Contour Plot of Cost Function J(w, b)", fontsize=14, fontweight="bold")
plt.xlabel("Weight (w)", fontsize=12)
plt.ylabel("Bias (b)", fontsize=12)
plt.legend(frameon=True)
plt.grid(True, linestyle=":", alpha=0.5)
plt.tight_layout()
plt.show()
```
