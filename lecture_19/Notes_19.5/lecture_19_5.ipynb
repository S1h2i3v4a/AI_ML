# Lecture 19.5: Operations on Box Plot
**Topics Covered:**
- Horizontal box plots (`vert=False`)
- Adding the Mean marker & line (`showmeans=True`, `meanline=True`)
- Adjusting whisker limits (`whis=2.0`)
- Comparing multiple datasets with `tick_labels` and `patch_artist`

```python
import matplotlib.pyplot as plt
import numpy as np

data = [7, 8, 5, 6, 9, 4, 10, 12, 15]

# Horizontal Box plot with mean indicator
plt.figure(figsize=(8, 4))
plt.boxplot(data, vert=False, showmeans=True, meanline=True, whis=2.0)
plt.title('Horizontal Box Plot (whis=2.0, meanline=True)', fontsize=13, fontweight='bold')
plt.xlabel('Measured Values', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Multiple Datasets Comparison

```python
np.random.seed(42)
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.figure(figsize=(8, 5))
box = plt.boxplot([group1, group2], tick_labels=['Group 1', 'Group 2'], patch_artist=True, showmeans=True)

# Add aesthetic colors
colors = ['#7293CB', '#E1974C']
for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.8)

plt.title('Performance Comparison: Group 1 vs Group 2', fontsize=14, fontweight='bold')
plt.ylabel('Performance Score', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
