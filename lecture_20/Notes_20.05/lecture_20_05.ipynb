# Day 20 - Lecture 20.5: Types of Events

**Topic:** Types of Events

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

sample_space = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {1, 3, 5}
C = {1, 2, 3, 4}

def prob(e): return len(e) / len(sample_space)

print(f"P(A) = {prob(A)}, P(B) = {prob(B)}, P(C) = {prob(C):.3f}")
print(f"Disjoint A & B: {prob(A.intersection(B)) == 0}")
print(f"Independence A & B: {np.isclose(prob(A)*prob(B), prob(A.intersection(B)))}")
print(f"Independence A & C: {np.isclose(prob(A)*prob(C), prob(A.intersection(C)))}")
```
