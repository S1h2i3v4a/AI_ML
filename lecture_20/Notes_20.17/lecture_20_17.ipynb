This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np
from scipy.stats import norm

x = np.linspace(-5, 5, 1000)
pdf = norm.pdf(x, 0, 1)
print(f"Trapezoidal Area: {np.trapezoid(pdf, x):.4f}")
```
