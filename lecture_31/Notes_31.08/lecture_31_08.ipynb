# Lecture 31.08: Choosing Optimal K (Elbow & Silhouette Code)
Standalone Jupyter Notebook demonstrating WCSS Elbow Curve and Silhouette Score evaluation.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate synthetic data
X, y = make_blobs(n_samples=500, centers=4, n_features=2, cluster_std=1.0, random_state=42)
```

### Choose our K value - elbow; silhouette score

```python
# Elbow Method

wcss = []
for k in range(1, 21):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
    
```

```python
sns.lineplot(x=range(1, 21), y=wcss, marker='o')
```

### Silhouette Score

```python
from sklearn.metrics import silhouette_score
```

```python
ss = []

for k in range(2, 21):
    kmeans = KMeans(n_clusters=k)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)

    ss.append(score)
```

```python
# plot - K & ss

sns.lineplot(x=range(2, 21), y=ss, marker='o')
```
