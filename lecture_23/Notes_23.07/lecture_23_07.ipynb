# Starting with Linear Regression (Hypothesis & Geometric Interpretation)

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(10, 60, 50)
y = 250 * x + 2000 + np.random.randn(50) * 1200

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', alpha=0.6, label='Data (x, y)')
plt.plot(x, 250*x + 2000, 'r-', lw=2.5, label='y_hat = 250*x + 2000')
plt.title('Linear Regression: Hypothesis Line')
plt.xlabel('Feature x')
plt.ylabel('Target y')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
