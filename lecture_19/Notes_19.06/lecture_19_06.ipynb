# Lecture 19.6: Stack Plots
**Topics Covered:**
- Concept of Stack Plots (Stacked Area Charts)
- Part-to-whole analysis across time
- Comparison with Pie Charts
- Case study: Multi-channel website traffic analysis

```python
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
direct = [50, 60, 70, 80, 90, 100, 110]
organic = [30, 40, 50, 55, 60, 70, 80]
social = [20, 25, 30, 35, 40, 50, 60]

plt.figure(figsize=(9, 6))
plt.stackplot(
    days, direct, organic, social,
    labels=['Direct', 'Organic', 'Social'],
    colors=['#386cb0', '#7fc97f', '#fdc086'],
    alpha=0.85
)

plt.title('Website Traffic Composition Over a Week', fontsize=14, fontweight='bold')
plt.xlabel('Days', fontsize=12)
plt.ylabel('Number of Visitors', fontsize=12)
plt.legend(loc='upper left', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```
