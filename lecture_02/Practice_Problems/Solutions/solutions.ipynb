### 📌 Case Study 1: Numerical Stability & IEEE-754 Precision Mechanics


```python
# Case 1 Numerical Benchmark
x_test = 1e16
naive_res = np.sqrt(x_test + 1.0) - np.sqrt(x_test)
stable_res = 1.0 / (np.sqrt(x_test + 1.0) + np.sqrt(x_test))

print(f"--- CASE 1 RESULTS AT x = {x_test:.0e} ---")
print(f"Naive Formulation (Cancellation Collapse): {naive_res}")
print(f"Stable Conjugate Formulation:              {stable_res:.16e}")
print(f"Machine Epsilon eps_mach:                  {np.finfo(np.float64).eps}")
```

### 📌 Case Study 2: CPython Integer Memory Optimization & Object Identity


```python
# Case 2 CPython small_ints Benchmark
a1, b1 = int("256"), int("256")
a2, b2 = int("257"), int("257")

print(f"--- CASE 2 CPYTHON CACHE PROOF ---")
print(f"n = 256 (Inside Cache [-5, 256]): a is b -> {a1 is b1} (Same Memory: {id(a1) == id(b1)})")
print(f"n = 257 (Outside Cache):          a is b -> {a2 is b2} (Distinct Heap Allocations)")
print(f"Small Integer Size: {sys.getsizeof(256)} bytes")
print(f"Arbitrary Precision 2^1000 Size: {sys.getsizeof(2**1000)} bytes")
```

### 📌 Case Study 3: Short-Circuit Logical Evaluation & Branch Prediction


```python
# Case 3 Short-Circuit Efficiency
calls = {"fast": 0, "expensive": 0}

def fast_check(v):
    calls["fast"] += 1
    return v > 0

def expensive_check(v):
    calls["expensive"] += 1
    return v == 42

inputs = [-10, -5, 0, 10, 42]
for item in inputs:
    # Short-circuit guarded evaluation
    _ = fast_check(item) and expensive_check(item)

print(f"--- CASE 3 SHORT-CIRCUIT SAVINGS ---")
print(f"Total Items:            {len(inputs)}")
print(f"Fast Predicate Calls:   {calls['fast']}")
print(f"Expensive Calls:        {calls['expensive']} (Bypassed {len(inputs) - calls['expensive']} calls)")
print(f"Bypass Efficiency Rate: {(1.0 - calls['expensive'] / len(inputs)) * 100:.1f}%")
```
