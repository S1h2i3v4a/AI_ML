# Lecture 19.15: Relational Plots in Seaborn - Heatmaps
**Topics Covered:**
- What is a Heatmap and Matrix plot
- Transforming data with Pandas `.pivot()`
- Customizing heatmaps (`cmap`, `annot`, `fmt`, `linewidths`)
- Pearson correlation matrix visualization (`df.corr()`)

```python
import seaborn as sns
import matplotlib.pyplot as plt

flights = sns.load_dataset('flights')
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')

plt.figure(figsize=(11, 7))
sns.heatmap(flights_pivot, cmap='coolwarm', annot=True, fmt='d', linewidths=0.4)
plt.title('Monthly Flight Passengers Matrix (1949 - 1960)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Feature Correlation Matrix Heatmap

```python
tips = sns.load_dataset('tips')
corr = tips.select_dtypes(include=['float64', 'int64']).corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap='vlag', vmin=-1, vmax=1, center=0, square=True)
plt.title('Correlation Heatmap', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```
