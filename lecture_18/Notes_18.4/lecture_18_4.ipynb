# Lecture 18.4: Important Plot Methods
**Topics Covered:**
- Essential Pyplot methods: `title`, `xlabel`, `ylabel`, `grid`, `show`
- Case study: Oscar Winner Movies Revenue analysis

```python
import matplotlib.pyplot as plt

oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.figure(figsize=(8, 5))
plt.plot(oscar_years, oscar_revenue, marker='o', color='royalblue', linewidth=2)
plt.title('Oscar Years vs Revenue (in $M)', fontsize=14, fontweight='bold')
plt.xlabel('Years', fontsize=12)
plt.ylabel('Revenue (in $M)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
