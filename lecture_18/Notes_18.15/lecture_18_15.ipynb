# Lecture 18.15: Add Annotations
**Topics Covered:**
- Labeling individual scatter observations with `plt.annotate()`
- Offset strategies to prevent marker obstruction
- Margin expansion using `plt.xlim()`

```python
import matplotlib.pyplot as plt

people = ['Person A', 'Person B', 'Person C', 'Person D', 'Person E', 
          'Person F', 'Person G', 'Person H', 'Person I', 'Person J']
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.figure(figsize=(10, 6))
plt.scatter(age, blood_pressure, c=blood_pressure, cmap='OrRd', s=120, edgecolor='black')
plt.colorbar(label='BP')

# Annotate each point with patient name
for i in range(len(age)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i]), xytext=(age[i] + 0.6, blood_pressure[i] + 0.4))

# Margin adjustment to avoid clipping
plt.xlim(min(age) - 2, max(age) + 10)
plt.title('Age vs Blood Pressure with Annotations', fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Blood Pressure')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
