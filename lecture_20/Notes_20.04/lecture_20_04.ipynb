**Topic:** Solving Foundation Probability Practice Problems

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import math
from itertools import product

die = list(range(1, 7))
s_space = list(product(die, die))
favorable = [p for p in s_space if sum(p) in (7, 8)]
print(f"Dice Sum 7 or 8: {len(favorable)} / {len(s_space)} = {len(favorable)/len(s_space):.4f}")

c_total = math.comb(52, 2)
c_kings = math.comb(4, 2)
print(f"2 Kings from 52 cards: {c_kings} / {c_total} = {c_kings/c_total:.6f} (Exact: 1/221 = {1/221:.6f})")
```
