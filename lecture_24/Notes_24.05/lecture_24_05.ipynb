# Underfitting (High Bias & Oversimplification)

Interactive notebook for Module 24.05: Underfitting (High Bias & Oversimplification).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Generate non-linear parabolic data
np.random.seed(42)
X = np.linspace(-3, 3, 30)
y = 0.8 * (X ** 2) + np.random.normal(0, 0.5, size=len(X))

# Fit an underfitted linear model to quadratic data
linear_model = LinearRegression()
linear_model.fit(X[:, np.newaxis], y)
y_pred = linear_model.predict(X[:, np.newaxis])

train_mse = mean_squared_error(y, y_pred)
print(f"Underfitted Linear Model Training MSE: {train_mse:.4f}")

# Underfitting: High error across both training and validation

```
