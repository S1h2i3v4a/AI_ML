# Lecture 19.10: Introduction to Seaborn
**Topics Covered:**
- What is Seaborn and why use it over plain Matplotlib
- Setting global themes (`sns.set_theme()`)
- Inspecting built-in datasets (`sns.get_dataset_names()`)
- Loading and inspecting the `tips` dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Verify version and inspect available datasets
print('Seaborn Version:', sns.__version__)
datasets = sns.get_dataset_names()
print(f'Total available datasets: {len(datasets)}')
print('Sample datasets:', datasets[:8])

# Configure default theme
sns.set_theme(style='whitegrid', palette='deep')

# Load sample dataset
tips = sns.load_dataset('tips')
tips.head()
```
