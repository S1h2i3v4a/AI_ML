# Regularization: Lasso Regression (L1 Norm & Feature Selection)

Interactive notebook for Module 24.08: Regularization: Lasso Regression (L1 Norm & Feature Selection).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso

# Synthetic data with 5 features (only 2 are truly predictive)
np.random.seed(42)
m = 100
X = np.random.randn(m, 5)
# True relationship: y = 4*x1 - 3*x2 + noise (x3, x4, x5 are useless noise)
y = 4 * X[:, 0] - 3 * X[:, 1] + np.random.randn(m) * 0.5

# Fit Lasso with moderate penalty
lasso = Lasso(alpha=0.5)
lasso.fit(X, y)

print("True Coefficients:    [4.0, -3.0, 0.0, 0.0, 0.0]")
print("Learned Coefficients: ", lasso.coef_.round(3))
print("Intercept:            ", round(lasso.intercept_, 3))
print("Notice how noise features (x3, x4, x5) were driven exactly to 0.0!")

```
