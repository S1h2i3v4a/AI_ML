# Day 21 - Lecture 21.8: Vector Addition & Geometric Laws

**Topic:** Vector Addition & Geometric Laws

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

u = np.array([3.0, 1.0])
v = np.array([1.0, 4.0])
w = u + v

print(f"u: {u}, v: {v}")
print(f"Resultant u + v: {w}")
print(f"Norm of resultant: {np.linalg.norm(w):.3f}")
```
