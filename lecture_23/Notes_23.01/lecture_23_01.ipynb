Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
X = np.array([1.0, 1.5, 2.0, 2.5, 3.0])
y = np.array([2.1, 2.9, 4.1, 5.0, 5.9])

# Fit simple line y = w*x + b
X_mean, y_mean = np.mean(X), np.mean(y)
w = np.sum((X - X_mean)*(y - y_mean)) / np.sum((X - X_mean)**2)
b = y_mean - w * X_mean

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', s=70, label='Training Data (E)')
plt.plot(X, w*X + b, 'r-', lw=2, label=f'Learned Model: y = {w:.2f}x + {b:.2f}')
plt.title('Machine Learning: Finding the Mapping Function f(x)', fontsize=12, fontweight='bold')
plt.xlabel('Input Feature (x)')
plt.ylabel('Target Value (y)')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
