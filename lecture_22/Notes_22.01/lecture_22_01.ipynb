# Introduction to Calculus for AI & Continuous Change

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

# Continuous function
x = np.linspace(-5, 5, 500)
y = x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y, label='y = x^2', color='blue', lw=2)
plt.axvline(0, color='black', ls='--', lw=1)
plt.axhline(0, color='black', ls='--', lw=1)

# Tangent line at x=2 (f'(2) = 4, tangent: y = 4*(x-2) + 4)
x_tangent = np.linspace(0, 4, 100)
y_tangent = 4 * (x_tangent - 2) + 4
plt.plot(x_tangent, y_tangent, 'r--', label='Tangent at x=2 (Slope=4)', lw=2)
plt.scatter([2], [4], color='red', s=70, zorder=5)

plt.title('Calculus: Instantaneous Rate of Change (Tangent Line)', fontsize=12, fontweight='bold')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
