# Day 20 - Lecture 20.19: Uniform Distribution

**Topic:** Uniform Distribution

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

a, b = 0.0, 10.0
vals = np.random.uniform(a, b, 500000)

print(f"Mean:     {np.mean(vals):.3f}")
print(f"Variance: {np.var(vals):.3f} (Exact: {((b-a)**2)/12:.3f})")
```
