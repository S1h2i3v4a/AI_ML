```python
# Verification of Rule-Based vs Data-Driven Paradigm
import numpy as np

# Traditional Rule-Based: Handcrafted function
def rule_based_celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

# Machine Learning Paradigm: Learning weights from data
# y = w * x + b
X = np.array([-40, 0, 10, 20, 30, 40], dtype=float)
y = np.array([-40, 32, 50, 68, 86, 104], dtype=float)

# Fit linear model directly using Ordinary Least Squares
X_b = np.c_[np.ones(len(X)), X]
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

print(f"Rule-Based Target: w = 1.80, b = 32.00")
print(f"ML Learned Parameters: w = {theta[1]:.4f}, b = {theta[0]:.4f}")
print("Successfully verified ML parameter recovery from pure data!")
```
