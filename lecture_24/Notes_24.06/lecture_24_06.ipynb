# How to Fix Underfitting & Overfitting (Practical Engineering Playbook)

Interactive notebook for Module 24.06: How to Fix Underfitting & Overfitting (Practical Engineering Playbook).


```python
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge, LinearRegression
from sklearn.metrics import mean_squared_error

# Generate synthetic data with noise
np.random.seed(42)
X_train = np.sort(np.random.uniform(0, 1, 20))
y_train = np.sin(2 * np.pi * X_train) + np.random.normal(0, 0.2, 20)

X_val = np.sort(np.random.uniform(0, 1, 30))
y_val = np.sin(2 * np.pi * X_val) + np.random.normal(0, 0.2, 30)

# Solution to Overfitting: Adding Ridge Regularization to high-degree polynomial
# Unregularized polynomial (Overfitting)
model_overfit = make_pipeline(PolynomialFeatures(degree=8), LinearRegression())
model_overfit.fit(X_train[:, np.newaxis], y_train)

# Regularized polynomial (Fixed with Ridge alpha=0.1)
model_fixed = make_pipeline(PolynomialFeatures(degree=8), Ridge(alpha=0.1))
model_fixed.fit(X_train[:, np.newaxis], y_train)

mse_overfit_val = mean_squared_error(y_val, model_overfit.predict(X_val[:, np.newaxis]))
mse_fixed_val = mean_squared_error(y_val, model_fixed.predict(X_val[:, np.newaxis]))

print(f"Unregularized Model Validation MSE (Overfitted): {mse_overfit_val:.4f}")
print(f"Ridge Regularized Model Validation MSE (Fixed):     {mse_fixed_val:.4f}")

```
