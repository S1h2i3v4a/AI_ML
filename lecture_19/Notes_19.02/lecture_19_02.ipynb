# Lecture 19.2: Multiple Datasets on Histogram
**Topics Covered:**
- Comparing two or more distributions on the same axes
- Use of `alpha` for transparency and visibility
- Adding descriptive legends
- Case study: Legitimate vs Fraudulent Transaction amounts

```python
import matplotlib.pyplot as plt
import numpy as np

# Transaction datasets from lecture
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

plt.figure(figsize=(10, 6))
plt.hist(fraud_transactions, bins=10, label='Fraudulent', color='crimson', alpha=0.55, edgecolor='black')
plt.hist(legit_transactions, bins=10, label='Legitimate', color='seagreen', alpha=0.55, edgecolor='black')

plt.xlabel('Transaction Amount ($)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Transaction Amount Distribution: Legitimate vs Fraudulent', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Side-by-Side Histogram Comparison
Instead of overlapping, we can pass both arrays as a list: `plt.hist([data1, data2])`.

```python
np.random.seed(42)
male_heights = np.random.normal(175, 8, 300)
female_heights = np.random.normal(162, 7, 300)

plt.figure(figsize=(9, 5))
plt.hist([male_heights, female_heights], bins=15, label=['Male Heights (cm)', 'Female Heights (cm)'],
         color=['steelblue', 'coral'], edgecolor='black')
plt.title('Side-by-Side Height Distribution Comparison', fontsize=14, fontweight='bold')
plt.xlabel('Height (cm)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```
