# Day 20 - Lecture 20.6: The Complementary Rule

**Topic:** The Complementary Rule

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

n_gpus = 10
p_fail = 0.05
p_safe = 1.0 - p_fail

p_zero = p_safe ** n_gpus
p_at_least_one = 1.0 - p_zero

sims = np.random.binomial(n=n_gpus, p=p_fail, size=1_000_000)
p_sim = np.mean(sims >= 1)

print(f"Analytical P(At Least 1 Fail): {p_at_least_one:.4f}")
print(f"Monte Carlo Simulation:       {p_sim:.4f}")
```
