# Day 20 - Lecture 20.3: Classical vs. Empirical Probability

**Topic:** Classical vs. Empirical Probability

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

np.random.seed(42)
n_flips = 10000

tosses = np.random.binomial(1, 0.5, size=n_flips)
cumulative_heads = np.cumsum(tosses)
trial_numbers = np.arange(1, n_flips + 1)
empirical_p = cumulative_heads / trial_numbers

for step in [10, 50, 100, 1000, 5000, 10000]:
    print(f"Trials = {step:5d} -> Empirical P(Heads) = {empirical_p[step-1]:.4f}")
```
