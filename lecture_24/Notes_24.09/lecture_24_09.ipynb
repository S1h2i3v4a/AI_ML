# Regularization: Ridge Regression (L2 Norm & Weight Shrinkage)

Interactive notebook for Module 24.09: Regularization: Ridge Regression (L2 Norm & Weight Shrinkage).


```python
import numpy as np
from sklearn.linear_model import Ridge

# Multicollinear dataset
np.random.seed(42)
m = 50
x1 = np.random.randn(m)
x2 = x1 + np.random.normal(0, 0.01, m) # x2 is almost identical to x1!
y = 3 * x1 + 2 * x2 + np.random.randn(m) * 0.5
X = np.c_[x1, x2]

# Ridge Regression handles extreme correlation smoothly
ridge = Ridge(alpha=10.0)
ridge.fit(X, y)

print("Ridge Coefficients with severe correlation:")
print(f"w1: {ridge.coef_[0]:.3f}, w2: {ridge.coef_[1]:.3f}")
print("Ridge evenly distributes and shrinks weights across correlated features!")

```
