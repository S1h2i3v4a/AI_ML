# Lecture 18.14: Customizations on Scatter Plots
**Topics Covered:**
- Dynamic conditional coloring via list comprehension
- Sizing markers dynamically (`s=sizes`)
- Continuous color scales (`c=values`, `cmap='OrRd'`)
- Adding and styling colorbars (`plt.colorbar()`)

```python
import matplotlib.pyplot as plt

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

# 1. Conditional styling with marker triangles
colors = ['green' if bp < 135 else 'red' for bp in blood_pressure]
sizes = blood_pressure

plt.figure(figsize=(8, 5))
plt.scatter(age, blood_pressure, color=colors, s=sizes, marker='^', alpha=0.6, edgecolor='black')
plt.title('Age vs Blood Pressure (Categorical Threshold < 135)', fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Blood Pressure')
plt.grid(True)
plt.show()

# 2. Continuous Colormap
plt.figure(figsize=(8, 5))
plt.scatter(age, blood_pressure, c=blood_pressure, cmap='OrRd', s=100, edgecolor='black')
plt.colorbar(label='BP Scale')
plt.title('Age vs BP with Colormap (OrRd)', fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Blood Pressure')
plt.show()
```
