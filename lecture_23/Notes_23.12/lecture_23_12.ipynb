# Summary of Linear Regression Foundations (Analytical vs Iterative Solvers)

Interactive notebook for Module 23.12: Summary of Linear Regression Foundations (Analytical vs Iterative Solvers).


```python
import numpy as np
import time

# Generate synthetic dataset with 10,000 samples and 5 features
np.random.seed(42)
m = 10000
d = 5
X = np.random.randn(m, d)
true_w = np.array([[2.5], [-1.2], [3.8], [0.5], [-4.0]])
y = X.dot(true_w) + 1.5 + np.random.randn(m, 1) * 0.2

# 1. Normal Equation
t0 = time.time()
X_b = np.c_[np.ones((m, 1)), X]
w_normal = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
t_normal = time.time() - t0

# 2. Batch Gradient Descent
t0 = time.time()
alpha = 0.05
iterations = 500
w_gd = np.zeros((d + 1, 1))

for _ in range(iterations):
    gradients = (1 / m) * X_b.T.dot(X_b.dot(w_gd) - y)
    w_gd -= alpha * gradients
t_gd = time.time() - t0

print(f"Normal Equation Time: {t_normal:.5f} s")
print(f"Normal Eq Intercept: {w_normal[0, 0]:.4f}, Slopes: {w_normal[1:, 0].round(3)}")
print(f"Gradient Descent Time: {t_gd:.5f} s")
print(f"GD Intercept: {w_gd[0, 0]:.4f}, Slopes: {w_gd[1:, 0].round(3)}")
```
