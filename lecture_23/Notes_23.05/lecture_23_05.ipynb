# Regression vs. Classification Tasks

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(1, 10, 30)
y_reg = 2.5 * X + np.random.randn(30) * 1.5
y_clf = (y_reg > 14).astype(int)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.scatter(X, y_reg, color='blue')
ax1.plot(X, 2.5 * X, 'r-', lw=2)
ax1.set_title('Regression: Continuous Output')
ax1.grid(True, linestyle=':')

ax2.scatter(X, y_clf, c=y_clf, cmap='bwr', s=70)
ax2.axvline(5.6, color='k', ls='--')
ax2.set_title('Classification: Discrete {0, 1} Output')
ax2.grid(True, linestyle=':')

plt.tight_layout()
plt.show()
```
