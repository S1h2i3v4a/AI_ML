# Types of Machine Learning: Supervised Learning

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
y = np.array([2.2, 3.9, 6.1, 7.9, 10.2])

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', s=80, label='Supervised Labeled Points (x_i, y_i)')
plt.plot(X, 2.0*X, 'r--', label='Model Hypothesis h(x) = 2.0*x')
for xi, yi in zip(X, y):
    plt.vlines(xi, 2.0*xi, yi, color='black', linestyle=':', label='Supervisory Error' if xi==1 else "")

plt.title('Supervised Learning: Features, Targets & Residual Errors', fontsize=12, fontweight='bold')
plt.xlabel('Feature x')
plt.ylabel('Target y')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
