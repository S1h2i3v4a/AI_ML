# What is the Best Fit Line? (Residuals & Ordinary Least Squares)

Interactive notebook for Module 23.08: What is the Best Fit Line? (Residuals & Ordinary Least Squares).


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic dataset
np.random.seed(42)
x = np.array([1.5, 2.3, 3.1, 4.0, 5.2, 6.1, 7.0, 8.4, 9.1, 10.0])
y = 2.5 * x + 4.0 + np.random.normal(0, 2.0, size=len(x))

# Calculate sample means
x_mean = np.mean(x)
y_mean = np.mean(y)

# Closed-form OLS formulas
cov_xy = np.sum((x - x_mean) * (y - y_mean))
var_x = np.sum((x - x_mean) ** 2)
w_opt = cov_xy / var_x
b_opt = y_mean - w_opt * x_mean

print(f"Optimal Slope (w): {w_opt:.4f}")
print(f"Optimal Intercept (b): {b_opt:.4f}")

# Predictions and residuals
y_hat = w_opt * x + b_opt
residuals = y - y_hat
sse = np.sum(residuals ** 2)
print(f"Sum of Squared Errors (SSE): {sse:.4f}")

# Plot data, best fit line, and residuals
plt.figure(figsize=(9, 6))
plt.scatter(x, y, color="#2563eb", s=80, label="Data Points", zorder=3)
x_line = np.linspace(1, 10.5, 100)
plt.plot(x_line, w_opt * x_line + b_opt, color="#dc2626", linewidth=2.5, label=f"OLS Best Fit: y = {w_opt:.2f}x + {b_opt:.2f}")

# Plot residual dashed lines
for xi, yi, yhi in zip(x, y, y_hat):
    plt.plot([xi, xi], [yi, yhi], color="#9333ea", linestyle="--", alpha=0.7, linewidth=1.5)

plt.scatter(x_mean, y_mean, color="#16a34a", s=140, marker="X", label=f"Centroid (x_bar, y_bar)", zorder=4)
plt.title("Ordinary Least Squares: Best Fit Line and Residuals", fontsize=14, fontweight="bold")
plt.xlabel("Feature x", fontsize=12)
plt.ylabel("Target y", fontsize=12)
plt.legend(frameon=True)
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
