# Day 21 - Lecture 21.11: The Cross Product

**Topic:** The Cross Product

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

u = np.array([1.0, 0.0, 0.0])
v = np.array([0.0, 1.0, 0.0])
w = np.cross(u, v)

print(f"u: {u}, v: {v}")
print(f"Cross Product u x v: {w}")
print(f"Area of parallelogram: {np.linalg.norm(w):.2f}")
```
