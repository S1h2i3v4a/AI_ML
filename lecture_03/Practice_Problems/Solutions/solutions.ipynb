## Solution 01: Tax Bracket Calculator

```python
def calculate_tax(salary: float) -> tuple[float, float]:
    """Calculates statutory tax rate percentage and total tax liability."""
    if salary < 0:
        raise ValueError("Salary cannot be negative")
    elif salary < 30000:
        rate = 0.05
    elif salary <= 70000:
        rate = 0.15
    else:
        rate = 0.25
    return rate * 100.0, salary * rate

# Verification
for s in [25000.0, 50000.0, 85000.0]:
    pct, tax = calculate_tax(s)
    print(f"Salary: ${s:9,.2f} | Rate: {pct:4.1f}% | Total Tax: ${tax:9,.2f}")
```

## Solution 02: Even Numbers in Range [a, b]

```python
def print_even_numbers(a: int, b: int) -> list[int]:
    """Returns all even numbers in [a, b] inclusive using step-2 range."""
    start, end = min(a, b), max(a, b)
    if (start & 1) != 0:
        start += 1
    evens = list(range(start, end + 1, 2))
    return evens

# Verification
print("Even numbers [4, 18]:", print_even_numbers(4, 18))
print("Even numbers [15, 3]:", print_even_numbers(15, 3))
```

## Solution 03: Print Digits of a Number N

```python
def extract_digits(n: int) -> list[int]:
    """Extracts decimal digits left-to-right via pure arithmetic."""
    temp = abs(n)
    if temp == 0:
        return [0]
    rev_digits = []
    while temp > 0:
        rev_digits.append(temp % 10)
        temp //= 10
    rev_digits.reverse()
    return rev_digits

# Verification
for val in [312, 9045, 0, 7]:
    print(f"Number {val:6d} -> Digits: {extract_digits(val)}")
```

## Solution 04: Count Number of Digits in N

```python
def count_digits(n: int) -> int:
    """Counts digits via successive floor division."""
    temp = abs(n)
    if temp == 0:
        return 1
    count = 0
    while temp > 0:
        count += 1
        temp //= 10
    return count

# Verification
for val in [312, 0, 100000, 7, -9876]:
    print(f"Value: {val:8d} -> Total Digits: {count_digits(val)}")
```

## Solution 05: Sum of Digits of N

```python
def sum_of_digits(n: int) -> int:
    """Computes sum of digits via while loop."""
    temp = abs(n)
    total_sum = 0
    while temp > 0:
        total_sum += temp % 10
        temp //= 10
    return total_sum

# Verification
for val in [312, 9045, 0, 999]:
    print(f"Value: {val:6d} -> Digit Sum: {sum_of_digits(val)}")
```

## Solution 07: Sentinel Loop Simulation

```python
def simulate_sentinel(feed: list[str]):
    for raw in feed:
        cleaned = raw.strip().lower()
        print(f"Input: '{raw}'")
        if cleaned == "quit":
            print("  -> Sentinel 'Quit' encountered. Halting loop.")
            break
        try:
            val = float(cleaned)
            classification = "Positive" if val > 0 else ("Negative" if val < 0 else "Zero")
            print(f"  -> Number {val} is {classification}.")
        except ValueError:
            print("  -> Invalid input! Please enter a valid number or 'Quit'.")

simulate_sentinel(["45", "-12.5", "0", "invalid_text", "108", "Quit"])
```

## Solution 08: Simple Calculator via Match...Case

```python
def calculator(a: float, b: float, operation: str) -> float:
    """Multi-way dispatch calculator via pattern matching."""
    match operation.strip():
        case '+':
            return a + b
        case '-':
            return a - b
        case '*':
            return a * b
        case '/':
            if b == 0:
                raise ZeroDivisionError("Cannot divide by zero!")
            return a / b
        case _:
            raise ValueError(f"Unknown operation: '{operation}'")

# Verification
for op in ['+', '-', '*', '/']:
    print(f"12 {op} 4 = {calculator(12, 4, op)}")
```

## Solution 09: Sub-Linear Primality Verifier

```python
def is_prime(n: int) -> bool:
    """Optimized primality check in O(sqrt(N)) time."""
    if n <= 1:
        return False
    if n in (2, 3):
        return True
    if (n & 1) == 0 or n % 3 == 0:
        return False
    limit = int(n**0.5)
    for d in range(5, limit + 1, 6):
        if n % d == 0 or n % (d + 2) == 0:
            return False
    return True

# Verification
for val in [2, 3, 4, 17, 25, 29, 49, 97]:
    print(f"Number {val:2d} is Prime? {is_prime(val)}")
```

## Solution 10: Number Guessing Game Engine

```python
class GuessingGame:
    def __init__(self, secret: int = 67):
        self.secret = secret
        self.attempts = 0
        self.won = False

    def guess(self, val: int) -> str:
        self.attempts += 1
        if val > self.secret:
            return "Too high"
        elif val < self.secret:
            return "Too low"
        else:
            self.won = True
            return "Correct!"

# Verification with binary search simulation
game = GuessingGame(secret=67)
low, high = 1, 100
while not game.won:
    mid = (low + high) // 2
    res = game.guess(mid)
    print(f"Attempt #{game.attempts}: Guess {mid:3d} -> {res}")
    if res == "Too high":
        high = mid - 1
    elif res == "Too low":
        low = mid + 1
print(f"Won in {game.attempts} guesses!")
```

## Systems Case Study 2: Loop Speed Benchmark

```python
import time

def run_benchmarks(n: int = 50000):
    # while
    t0 = time.perf_counter()
    s = 0
    i = 0
    while i < n:
        s += i
        i += 1
    tw = time.perf_counter() - t0

    # for
    t0 = time.perf_counter()
    s = 0
    for i in range(n):
        s += i
    tf = time.perf_counter() - t0

    # C-accumulator
    t0 = time.perf_counter()
    s = sum(range(n))
    tc = time.perf_counter() - t0

    return tw, tf, tc

tw, tf, tc = run_benchmarks(50000)
print(f"while loop:   {tw*1000:.3f} ms")
print(f"for loop:     {tf*1000:.3f} ms ({(tw/tf):.1f}x faster)")
print(f"sum(range()): {tc*1000:.3f} ms ({(tw/tc):.1f}x faster)")
```

## Systems Case Study 3: Recursion Depth Limits

```python
import sys

print("CPython Default Recursion Limit:", sys.getrecursionlimit())

def recurse_down(depth: int, max_d: int = 5):
    if depth >= max_d:
        return "Base Case Reached"
    return recurse_down(depth + 1, max_d)

print(recurse_down(0, 5))
```
