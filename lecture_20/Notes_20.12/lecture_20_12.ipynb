# Day 20 - Lecture 20.12: Bayes' Theorem

**Topic:** Bayes' Theorem

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

p_prior = np.array([0.50, 0.30, 0.20])
p_likelihood = np.array([0.01, 0.02, 0.05])
p_evidence = np.sum(p_prior * p_likelihood)
p_posterior = (p_prior * p_likelihood) / p_evidence

print(f"Evidence P(D) = {p_evidence:.4f}")
print(f"Posterior P(Factory | Defect) = {p_posterior}")
```
