# Day 20 - Lecture 20.20: Normal Distribution

**Topic:** Normal Distribution

This interactive notebook contains reference code and computational experiments for this lecture.

```python
from scipy.stats import norm

mu = 70
sigma = 10

p1 = norm.cdf(mu + sigma, mu, sigma) - norm.cdf(mu - sigma, mu, sigma)
p2 = norm.cdf(mu + 2*sigma, mu, sigma) - norm.cdf(mu - 2*sigma, mu, sigma)
p3 = norm.cdf(mu + 3*sigma, mu, sigma) - norm.cdf(mu - 3*sigma, mu, sigma)

print(f"1-Sigma Region: {p1*100:.2f}%")
print(f"2-Sigma Region: {p2*100:.2f}%")
print(f"3-Sigma Region: {p3*100:.2f}%")
```
