# Differentiation & Instantaneous Rate of Change

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 5, 200)
y = x**2

# Point x0 = 2
x0 = 2
y0 = x0**2
slope = 2 * x0

# Tangent line
x_tan = np.linspace(0.5, 3.5, 100)
y_tan = slope * (x_tan - x0) + y0

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-', label='f(x) = x^2', lw=2)
plt.plot(x_tan, y_tan, 'r--', label='Tangent at x=2 (Slope=4)', lw=2)
plt.scatter([x0], [y0], color='red', s=80, zorder=5)
plt.title('Differentiation: Slope of Tangent Line')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
