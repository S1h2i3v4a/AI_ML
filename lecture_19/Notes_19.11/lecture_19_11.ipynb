# Lecture 19.11: Creating Plots with Seaborn
**Topics Covered:**
- Semantic dimensions: `hue`, `style`, `size`
- Figure-level plotting with `sns.relplot()`
- Faceting grids across categorical variables (`col='time'`)
- Line generation using `kind='line'`

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.set_theme(style='white')

# Multi-variable scatter with faceting
g = sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='time',
    hue='smoker',
    style='smoker',
    size='size',
    sizes=(30, 200),
    palette='Set1'
)
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle('Total Bill vs Tip by Time, Smoker, and Table Size', fontsize=14, fontweight='bold')
plt.show()
```
