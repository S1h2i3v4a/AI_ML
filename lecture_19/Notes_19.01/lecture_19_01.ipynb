# Lecture 19.1: Histograms in Matplotlib
**Topics Covered:**
- What is a Histogram and when to use it
- Basic `plt.hist()` parameters (`bins`, `edgecolor`, `color`, `alpha`)
- Equal vs Custom Bin intervals
- Continuous data distribution visualization

```python
import matplotlib.pyplot as plt
import numpy as np

# Seed for reproducible results
np.random.seed(0)

# Generate simulated exam scores (mean=70, std=10, 100 students)
scores = np.random.normal(loc=70, scale=10, size=100)

plt.figure(figsize=(8, 5))
plt.hist(scores, bins=10, color='skyblue', edgecolor='black', alpha=0.85)
plt.title('Distribution of Exam Scores (10 Bins)', fontsize=14, fontweight='bold')
plt.xlabel('Marks', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Custom Bins Example
Often in business and grading systems, intervals are unequal (e.g., Fail < 33, Second Div 33-60, First Div 60-80, Distinction 80-100).

```python
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

# Define custom bin edges
bins = [0, 33, 60, 80, 100]

plt.figure(figsize=(8, 5))
plt.hist(marks, bins=bins, color='lightgreen', edgecolor='black', linewidth=1.2)
plt.title('Distribution of Marks (Custom Grade Bins)', fontsize=14, fontweight='bold')
plt.xlabel('Grade Range', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.xticks(bins)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
