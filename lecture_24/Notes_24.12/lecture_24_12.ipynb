# ElasticNet Overview (Hybrid Regularization)

Interactive notebook for Module 24.12: ElasticNet Overview (Hybrid Regularization).


```python
from sklearn.linear_model import ElasticNet
import numpy as np

# Synthetic grouped data
np.random.seed(42)
m = 100
# Group 1: highly correlated predictors
z1 = np.random.randn(m)
x1 = z1 + np.random.normal(0, 0.05, m)
x2 = z1 + np.random.normal(0, 0.05, m)
# Group 2: noise
x3 = np.random.randn(m)
x4 = np.random.randn(m)

X = np.c_[x1, x2, x3, x4]
y = 3 * z1 + np.random.randn(m) * 0.5

# Fit ElasticNet
elastic = ElasticNet(alpha=0.5, l1_ratio=0.5)
elastic.fit(X, y)

print("ElasticNet Coefficients:")
for i, c in enumerate(elastic.coef_):
    print(f"  Feature x{i+1}: {c:.3f}")
print("Notice how correlated group (x1, x2) is preserved together while noise is shrunk!")

```
