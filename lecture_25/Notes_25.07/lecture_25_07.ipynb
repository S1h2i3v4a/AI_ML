# Notes_25.07: Types of Naive Bayes Classifiers\nGaussianNB for continuous features, MultinomialNB for discrete counts, and BernoulliNB for binary indicators.

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
import numpy as np

# Continuous features -> GaussianNB
X_cont = np.array([[1.2, 5.4], [2.1, 4.8], [5.6, 1.2], [6.2, 0.9]])
y = [0, 0, 1, 1]
gnb = GaussianNB().fit(X_cont, y)
print("GaussianNB Classes:", gnb.classes_)
```
