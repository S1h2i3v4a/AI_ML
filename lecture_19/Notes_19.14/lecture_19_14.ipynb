# Lecture 19.14: Distribution Plots in Seaborn
**Topics Covered:**
- `sns.histplot()` vs `plt.hist()`
- Kernel Density Estimation (`kde=True`)
- Inspecting the Palmer Penguins dataset
- Comparing multi-class distributions with `hue`

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset('penguins')
print(penguins.head())

# 1. Basic histogram with KDE
plt.figure(figsize=(9, 5))
sns.histplot(data=penguins, x='body_mass_g', bins=30, kde=True, color='teal')
plt.title('Penguin Body Mass Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Body Mass (g)')
plt.show()

# 2. Multi-species comparison
plt.figure(figsize=(9, 5))
sns.histplot(data=penguins, x='body_mass_g', hue='species', kde=True, element='step')
plt.title('Body Mass by Species', fontsize=14, fontweight='bold')
plt.show()
```
