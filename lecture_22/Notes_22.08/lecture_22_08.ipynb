# Finding Minima & Maxima (First/Second Derivative Tests & Gradient Descent)

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

w_vals = np.linspace(-2, 8, 200)
loss_vals = (w_vals - 3)**2 + 4

# Run gradient descent steps
w = 8.0
history = [w]
for _ in range(8):
    w = w - 0.2 * (2 * (w - 3))
    history.append(w)

history = np.array(history)

plt.figure(figsize=(8, 5))
plt.plot(w_vals, loss_vals, 'b-', label='L(w) = (w - 3)^2 + 4', lw=2)
plt.plot(history, (history - 3)**2 + 4, 'ro-', label='Gradient Descent Steps', markersize=6)
plt.title('Optimization: Gradient Descent Steps to Minima')
plt.xlabel('Weight w')
plt.ylabel('Loss L(w)')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
