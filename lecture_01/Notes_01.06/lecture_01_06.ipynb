```python
# Diagnostic Script: Verify Python Environment Paths
import sys
import os

print("=== PYTHON RUNTIME PATH VERIFICATION ===")
print(f"Executable:  {sys.executable}")
print(f"Prefix:      {sys.prefix}")
print(f"Base Prefix: {sys.base_prefix}")

print("\nSystem PATH entries containing Python:")
paths = os.environ.get("PATH", "").split(os.pathsep)
py_paths = [p for p in paths if "python" in p.lower()]
if py_paths:
    for p in py_paths:
        print(f"  [PATH] {p}")
else:
    print("  [WARNING] No explicit 'python' keyword found in PATH strings.")
```
