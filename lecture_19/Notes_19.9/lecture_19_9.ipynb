# Lecture 19.9: Practice Task - Weekly Temperature Analysis
**Topics Covered:**
- End-to-end multi-city temperature visualization
- `axes.flatten()` vs nested loop iteration
- Global titles & axis labels (`fig.suptitle`, `fig.supxlabel`, `fig.supylabel`)
- Shared vs individual scales

```python
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
cities = ['New York', 'London', 'Delhi', 'Tokyo']

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes_flat = axes.flatten()
palette = ['royalblue', 'darkorange', 'forestgreen', 'firebrick']

for i, ax in enumerate(axes_flat):
    ax.plot(days, temperatures[i], marker='o', color=palette[i], linewidth=2.2, markersize=6)
    ax.set_title(cities[i], fontsize=13, fontweight='bold')
    ax.grid(True, linestyle=':', alpha=0.7)
    ax.set_ylim(15, 38) # Standardized scale

fig.suptitle('Weekly Temperature by City', fontsize=16, fontweight='bold')
fig.supylabel('Temperature (°C)', fontsize=13)
fig.supxlabel('Day', fontsize=13)

fig.tight_layout()
plt.show()
```
