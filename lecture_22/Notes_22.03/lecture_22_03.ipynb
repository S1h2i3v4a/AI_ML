# Composite Functions & Deep Neural Architectures

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 200)

# f(x) = x^2, g(x) = x + 1
f_of_g = (x + 1)**2
g_of_f = (x**2) + 1

plt.figure(figsize=(8, 5))
plt.plot(x, f_of_g, 'b-', label='f(g(x)) = (x + 1)^2', lw=2)
plt.plot(x, g_of_f, 'r--', label='g(f(x)) = x^2 + 1', lw=2)
plt.title('Non-Commutativity of Function Composition', fontsize=12, fontweight='bold')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
