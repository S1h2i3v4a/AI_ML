**Topic:** Core Foundations of Probability

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

sample_space = {1, 2, 3, 4, 5, 6}
event_even = {2, 4, 6}
event_odd = {1, 3, 5}

def prob(event, s=sample_space):
    return len(event.intersection(s)) / len(s)

print(f"Axiom 1 (Non-negativity): P(Even) = {prob(event_even)} >= 0 -> {prob(event_even) >= 0}")
print(f"Axiom 2 (Normalization): P(S) = {prob(sample_space)} == 1 -> {prob(sample_space) == 1.0}")
print(f"Axiom 3 (Additivity): P(Even U Odd) = {prob(event_even.union(event_odd))}, Sum = {prob(event_even) + prob(event_odd)}")
```
