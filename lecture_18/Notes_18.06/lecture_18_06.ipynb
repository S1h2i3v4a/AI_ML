# Lecture 18.6: Format Strings
**Topics Covered:**
- Shorthand `fmt = '[marker][linestyle][color]'` syntax
- Common marker, line, and color codes
- Explicit keyword arguments (`color`, `linestyle`, `linewidth`, `markersize`)

```python
import matplotlib.pyplot as plt

oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]
non_oscar_years = [2008, 2009, 2010, 2011, 2012]
non_oscar_revenue = [378, 2788, 829, 185, 275]

# Format Strings shorthand
plt.figure(figsize=(9, 5))
plt.plot(oscar_years, oscar_revenue, 'o-b', label='Oscar Winners')
plt.plot(non_oscar_years, non_oscar_revenue, '>--g', label='Non-Oscar Winners')
plt.legend()
plt.title('Format Strings Example')
plt.show()

# Explicit keyword arguments
plt.figure(figsize=(9, 5))
plt.plot(oscar_years, oscar_revenue, color='black', linestyle='-', marker='o', label='Oscar Winners')
plt.plot(non_oscar_years, non_oscar_revenue, color='red', linestyle='--', marker='>', linewidth=3, label='Non-Oscar Winners')
plt.legend()
plt.title('Explicit Keyword Styling')
plt.show()
```
