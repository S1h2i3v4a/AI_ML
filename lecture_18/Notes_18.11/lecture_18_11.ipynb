# Lecture 18.11: Multiple Datasets on Bar Chart
**Topics Covered:**
- Side-by-side grouped bar charts
- Using `np.arange()` for numeric coordinates
- Offset calculations (`x - width/2` and `x + width/2`)
- Restoring categorical labels with `plt.xticks()`

```python
import matplotlib.pyplot as plt
import numpy as np

oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))
width = 0.4

plt.figure(figsize=(10, 6))
plt.bar(x - width/2, oscar_revenue, width, label='Oscar Winners', color='royalblue', edgecolor='black')
plt.bar(x + width/2, non_oscar_revenue, width, label='Non-Oscar Winners', color='coral', edgecolor='black')

plt.xlabel('Years', fontsize=12)
plt.ylabel('Revenue (in $M)', fontsize=12)
plt.title('Yearly revenue of Oscar vs Non-Oscar movies', fontsize=14, fontweight='bold')
plt.xticks(x, years)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
