# Lecture 19.3: axvline in Histogram
**Topics Covered:**
- Adding vertical reference lines with `plt.axvline()`
- Setting passing score thresholds and average markers
- Line customization (`linestyle`, `linewidth`, `color`, `label`)
- Combining reference lines with legends and grids

```python
import matplotlib.pyplot as plt
import numpy as np

marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

bins = [0, 30, 70, 90, 100]

plt.figure(figsize=(9, 6))
plt.hist(marks, bins=bins, edgecolor='black', color='lightblue', alpha=0.85)

# Passing threshold line at x=33
plt.axvline(
    x=33,
    color='red',
    linestyle='--',
    linewidth=2.5,
    label='Pass Mark (33)'
)

# Average score line
mean_score = np.mean(marks)
plt.axvline(
    x=mean_score,
    color='navy',
    linestyle='-.',
    linewidth=2,
    label=f'Mean Score ({mean_score:.1f})'
)

plt.xlabel('Marks', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.title('Distribution of Marks of 30 Students with Cut-offs', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
