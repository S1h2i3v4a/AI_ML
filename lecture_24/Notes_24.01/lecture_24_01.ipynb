# Feature Engineering: Categorical Encoding

Interactive notebook for Module 24.01: Feature Engineering: Categorical Encoding.


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# Sample dataset with nominal and ordinal features
data = pd.DataFrame({
    "Device": ["Laptop", "Mobile", "Tablet", "Mobile", "Laptop"],
    "Rating": ["Poor", "Good", "Excellent", "Average", "Good"],
    "Price": [1200, 800, 500, 750, 1100]
})

print("Original Dataset:")
print(data)

# 1. Ordinal Encoding for Rating
rating_order = [["Poor", "Average", "Good", "Excellent"]]
ord_enc = OrdinalEncoder(categories=rating_order)
data["Rating_Encoded"] = ord_enc.fit_transform(data[["Rating"]])

# 2. One-Hot Encoding for Device
ohe = OneHotEncoder(sparse_output=False)
device_encoded = ohe.fit_transform(data[["Device"]])
device_df = pd.DataFrame(device_encoded, columns=ohe.get_feature_names_out(["Device"]))

final_df = pd.concat([data.drop(columns=["Device", "Rating"]), device_df], axis=1)
print("\nEncoded Dataset:")
print(final_df)

```
