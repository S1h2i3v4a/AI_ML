```python
import numpy as np

v = np.array([3.0, -4.0, 12.0])
mag = np.linalg.norm(v)
u = v / mag

print(f"Vector: {v}")
print(f"Magnitude: {mag:.2f}")
print(f"Unit Vector: {u}")
```
