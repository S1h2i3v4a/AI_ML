# Mathematical Functions & Real-World Transformations

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x_sqrt = np.linspace(0, 100, 1000)
y_sqrt = np.sqrt(x_sqrt)

x_sin = np.linspace(-2 * np.pi, 2 * np.pi, 500)
y_sin = np.sin(x_sin)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(x_sqrt, y_sqrt, 'm-', lw=2, label='y = sqrt(x)')
ax1.set_title('Square Root Function')
ax1.grid(True, linestyle=':')
ax1.legend()

ax2.plot(x_sin, y_sin, 'b-', lw=2, label='y = sin(x)')
ax2.axhline(0, color='k', ls='--')
ax2.axvline(0, color='k', ls='--')
ax2.set_title('Sine Function')
ax2.grid(True, linestyle=':')
ax2.legend()

plt.tight_layout()
plt.show()
```
