# Day 20 - Lecture 20.15: Mean, Median & Mode

**Topic:** Mean, Median & Mode

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

data = np.random.lognormal(mean=2.0, sigma=0.8, size=100000)
print(f"Sample Mean:   {np.mean(data):.3f}")
print(f"Sample Median: {np.median(data):.3f}")
```
