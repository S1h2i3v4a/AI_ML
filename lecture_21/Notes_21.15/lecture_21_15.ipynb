# Day 21 - Lecture 21.15: Eigenvalues & Eigenvectors

**Topic:** Eigenvalues & Eigenvectors

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

A = np.array([[4.0, 2.0], [2.0, 3.0]])
evals, evecs = np.linalg.eigh(A)

print("Eigenvalues:", evals)
print("Eigenvectors:
", evecs)
```
