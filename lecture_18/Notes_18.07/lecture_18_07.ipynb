# Lecture 18.7: Styling & Saving Plots
**Topics Covered:**
- Exploring built-in styles (`plt.style.available`, `plt.style.use()`)
- Sketch comic style with `with plt.xkcd():`
- Auto-margin adjustment with `plt.tight_layout()`
- Exporting high-res images with `plt.savefig()`

```python
import matplotlib.pyplot as plt

print('Available Styles count:', len(plt.style.available))
print('Sample styles:', plt.style.available[:6])

oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]
non_oscar_revenue = [378, 2788, 829, 185, 275]

# Hand-drawn XKCD style
with plt.xkcd():
    plt.figure(figsize=(8, 4.5))
    plt.plot(oscar_years, oscar_revenue, 'o-b', label='Oscar')
    plt.plot(oscar_years, non_oscar_revenue, '>--g', label='Non-Oscar')
    plt.xlabel('Years')
    plt.ylabel('Revenue ($M)')
    plt.legend()
    plt.tight_layout()
    plt.savefig('xkcd_demo.png', dpi=100)
    plt.show()

plt.style.use('default')
```
