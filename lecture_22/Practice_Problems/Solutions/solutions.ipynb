## 📌 Case 1 Solution: Gradient Descent Simulation

```python
def loss(w): return 2 * (w - 3)**2 + 4
def grad(w): return 4 * (w - 3)

etas = [0.1, 0.4, 0.52]
labels = ['Stable (η=0.1)', 'Oscillatory (η=0.4)', 'Divergent (η=0.52)']

for eta, lbl in zip(etas, labels):
    w = 8.0
    history = [w]
    for _ in range(10):
        w = w - eta * grad(w)
        history.append(w)
    print(f"{lbl} -> final w after 10 steps: {history[-1]:.4f}")
```
