# Day 21 - Lecture 21.10: The Dot Product (Inner Product)

**Topic:** The Dot Product (Inner Product)

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

u = np.array([1.0, 2.0, 3.0])
v = np.array([4.0, 5.0, 6.0])

dot = np.dot(u, v)
cos_sim = dot / (np.linalg.norm(u) * np.linalg.norm(v))

print(f"u . v = {dot}")
print(f"Cosine Similarity = {cos_sim:.4f}")
```
