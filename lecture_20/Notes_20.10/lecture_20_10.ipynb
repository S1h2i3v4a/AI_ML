# Day 20 - Lecture 20.10: Conditional Probability

**Topic:** Conditional Probability

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import pandas as pd

df = pd.DataFrame({
    'CVD_Yes': [90, 35],
    'CVD_No': [210, 665]
}, index=['Smoker', 'NonSmoker'])

total = df.values.sum()
p_cvd_given_smoker = df.loc['Smoker', 'CVD_Yes'] / df.loc['Smoker'].sum()
p_overall_cvd = df['CVD_Yes'].sum() / total

print(f"P(CVD | Smoker) = {p_cvd_given_smoker:.4f}")
print(f"P(Overall CVD)  = {p_overall_cvd:.4f}")
```
