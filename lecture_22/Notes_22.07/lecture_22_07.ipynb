# Differentiation Rules & Activation Function Derivatives

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 300)
sig = 1 / (1 + np.exp(-x))
d_sig = sig * (1 - sig)

plt.figure(figsize=(8, 5))
plt.plot(x, sig, 'b-', label='Sigmoid: σ(x)', lw=2)
plt.plot(x, d_sig, 'r--', label="Sigmoid Derivative: σ'(x)", lw=2)
plt.title('Sigmoid Function and its Derivative')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
