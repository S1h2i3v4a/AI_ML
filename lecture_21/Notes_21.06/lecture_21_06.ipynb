# Day 21 - Lecture 21.6: Practice Problems: Coordinate Geometry & Linear Boundaries

**Topic:** Practice Problems: Coordinate Geometry & Linear Boundaries

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

w = np.array([2.0, 5.0])
b = -10.0
p = np.array([3.0, 2.0])

score = np.dot(w, p) + b
dist = abs(score) / np.linalg.norm(w)
print(f"Point {p} -> Score = {score}, Distance = {dist:.4f}")
```
