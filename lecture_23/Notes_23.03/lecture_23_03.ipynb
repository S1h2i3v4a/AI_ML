# Types of Machine Learning: Unsupervised & Reinforcement Learning

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

np.random.seed(42)
X = np.vstack([
    np.random.randn(40, 2) + [3, 3],
    np.random.randn(40, 2) + [-3, -3]
])

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X)

plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, cmap='viridis', s=70, label='Data Points')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], color='red', marker='X', s=200, label='Discovered Centroids')
plt.title('Unsupervised Learning: K-Means Clustering on Unlabeled Data', fontsize=12, fontweight='bold')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
