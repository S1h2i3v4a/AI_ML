```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm

print('All statistical packages loaded successfully.')
```

## Case 1: Sequential Bayesian Updating Pipeline

```python
p_prior = 0.002
sens_s1, fpr_s1 = 0.85, 0.04
sens_s2, fpr_s2 = 0.90, 0.03

# Single-signal update
p_s1 = sens_s1 * p_prior + fpr_s1 * (1 - p_prior)
post_s1 = (sens_s1 * p_prior) / p_s1

# Sequential second-signal update
p_s2 = sens_s2 * post_s1 + fpr_s2 * (1 - post_s1)
post_s2 = (sens_s2 * post_s1) / p_s2

print(f'Prior: {p_prior:.4f}')
print(f'Post S1: {post_s1:.4f}')
print(f'Post S2: {post_s2:.4f}')
```

## Case 2: Clinical Trial Binomial Testing

```python
n = 50
p0 = 0.20
k_obs = 18

p_val_exact = binom.sf(k_obs - 1, n, p0)
mu = n * p0
sigma = np.sqrt(n * p0 * (1 - p0))
z = (k_obs - 0.5 - mu) / sigma
p_val_gauss = norm.sf(z)

print(f'Exact Binomial p-value: {p_val_exact:.6f}')
print(f'Gaussian Z-score:       {z:.4f}')
print(f'Gaussian p-value:       {p_val_gauss:.6f}')
```

## Case 3: Autonomous Vehicle LiDAR Gaussian Calibration

```python
sigma = 0.15
p_green = norm.cdf(1) - norm.cdf(-1)
p_yellow = (norm.cdf(2) - norm.cdf(-2)) - p_green
p_red = 2 * norm.sf(3)
p_cutoff = 2 * norm.sf(0.25 / sigma)

print(f'Green: {p_green:.4f}, Yellow: {p_yellow:.4f}, Red: {p_red:.6f}')
print(f'P(|err| > 0.25m): {p_cutoff:.4f}')
```
