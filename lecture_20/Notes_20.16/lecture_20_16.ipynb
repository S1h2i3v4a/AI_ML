# Day 20 - Lecture 20.16: Variance & Standard Deviation

**Topic:** Variance & Standard Deviation

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

x = np.array([10, 12, 14, 16, 18, 20])
mu = np.mean(x)
var = np.mean((x - mu)**2)
std = np.sqrt(var)

print(f"mu = {mu}, var = {var:.4f}, std = {std:.4f}")
```
