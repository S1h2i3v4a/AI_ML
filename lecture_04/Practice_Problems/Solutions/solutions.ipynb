## Solution 01: Palindrome String Check

```python
def is_palindrome(s: str, case_sensitive: bool = False) -> bool:
    """Two-pointer palindrome verification in O(N) time and O(1) space."""
    if not case_sensitive:
        s = s.lower()
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

# Verification
test_cases = ["radar", "Python", "racecar", "Madam", "level", "step on no pets"]
for word in test_cases:
    clean_word = "".join(c for c in word if c.isalnum())
    print(f"Word: '{word:20s}' | Palindrome: {is_palindrome(clean_word)}")
```

## Solution 02: Average of Integer List

```python
from typing import List, Optional

def compute_average(nums: List[int]) -> Optional[float]:
    """Calculates arithmetic average with zero division guard."""
    if not nums:
        return None
    return sum(nums) / len(nums)

# Verification
samples = [
    [10, 20, 30, 40, 50],
    [5, 12, 17, 24],
    [100],
    []
]
for s in samples:
    avg = compute_average(s)
    if avg is not None:
        print(f"Data: {s!s:25s} -> Average: {avg:.2f}")
    else:
        print(f"Data: {s!s:25s} -> Average: [Empty Sequence]")
```

## Solution 03: Merge and Sort Two Lists

```python
from typing import List

def merge_and_sort(list1: List[int], list2: List[int]) -> List[int]:
    """Concatenates two lists and sorts them via Timsort in O((N+M) log(N+M))."""
    merged = list1 + list2
    merged.sort()
    return merged

# Verification
l1 = [4, 1, 7, 12]
l2 = [9, 2, 5, 0]
res = merge_and_sort(l1, l2)
print(f"List 1: {l1}")
print(f"List 2: {l2}")
print(f"Merged and Sorted: {res}")
```

## Solution 04: Tuple Parity Partitioning

```python
from typing import Tuple

def partition_tuple(numbers: Tuple[int, ...]) -> Tuple[Tuple[int, ...], Tuple[int, ...]]:
    """Partitions tuple into (evens, odds) in O(N) time preserving order."""
    evens = tuple(x for x in numbers if (x & 1) == 0)
    odds = tuple(x for x in numbers if (x & 1) != 0)
    return evens, odds

# Verification
data = (10, 15, 22, 33, 40, 55, 62, 77, 88)
evens, odds = partition_tuple(data)
print(f"Source Tuple: {data}")
print(f"Even Numbers: {evens}")
print(f"Odd Numbers:  {odds}")
```

## Solution 05: Student Gradebook Hash Table

```python
from typing import Dict

class StudentGradebook:
    """Production hash table gradebook."""
    def __init__(self):
        self._records: Dict[str, str] = {}

    def add(self, name: str, grade: str) -> bool:
        norm = name.strip().title()
        if norm in self._records:
            return False
        self._records[norm] = grade.strip().upper()
        return True

    def update(self, name: str, grade: str) -> bool:
        norm = name.strip().title()
        if norm not in self._records:
            return False
        self._records[norm] = grade.strip().upper()
        return True

    def search(self, name: str) -> str:
        norm = name.strip().title()
        return self._records.get(norm, "NOT_FOUND")

    def display(self):
        print(f"{'Student Name':<20} | {'Grade':<8}")
        print("-" * 32)
        for name, grade in sorted(self._records.items()):
            print(f"{name:<20} | {grade:<8}")

# Verification
gb = StudentGradebook()
gb.add("Alice Smith", "A")
gb.add("Bob Johnson", "B+")
gb.add("Charlie Brown", "A-")
gb.display()
print("\nUpdating Bob Johnson to 'A'...")
gb.update("Bob Johnson", "A")
print("Search Bob Johnson:", gb.search("Bob Johnson"))
```

## Solution 06: Word to Length Dictionary

```python
from typing import List, Dict

def word_length_dict(words: List[str]) -> Dict[str, int]:
    """Dictionary comprehension for mapping word to its length."""
    return {w: len(w) for w in words}

# Verification
words_corpus = ["apple", "banana", "kiwi", "cherry", "pomegranate"]
mapping = word_length_dict(words_corpus)
print(f"Corpus: {words_corpus}")
print(f"Mapping: {mapping}")
```

## Solution 07: Whitespace Character Counter

```python
def count_whitespaces(text: str) -> int:
    """Counts whitespace characters in O(N) time and O(1) space."""
    return sum(1 for ch in text if ch.isspace())

# Verification
raw_text = "Machine Learning\tLecture 04\nAssignment Suite\r 2026"
cnt = count_whitespaces(raw_text)
print(f"Raw String: {repr(raw_text)}")
print(f"Total Whitespace Characters: {cnt}")
```

## Solution 08: Disjoint Set Verification

```python
from typing import List, Any

def are_disjoint(l1: List[Any], l2: List[Any]) -> bool:
    """Tests if two sequences share zero common elements."""
    return set(l1).isdisjoint(l2)

# Verification
a = [1, 2, 3, 4]
b = [5, 6, 7, 8]
c = [4, 9, 10]
print(f"A & B disjoint? {are_disjoint(a, b)}")
print(f"A & C disjoint? {are_disjoint(a, c)}")
```

## Solution 09: Sequence Duplicate Detector

```python
from typing import List, Set, Any

def extract_duplicates(items: List[Any]) -> Set[Any]:
    """Single-pass duplicate extractor using seen and duplicates sets."""
    seen, duplicates = set(), set()
    for item in items:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    return duplicates

# Verification
numbers = [1, 2, 3, 2, 4, 5, 1, 6, 2, 7, 5]
dups = extract_duplicates(numbers)
print(f"Sequence: {numbers}")
print(f"Duplicates Detected: {sorted(dups)}")
```

## Solution 10: Unique Character Extraction & Cardinality

```python
from typing import Tuple, Set

def extract_unique_chars(text: str) -> Tuple[Set[str], int]:
    """Extracts unique characters and cardinality."""
    unique_set = set(text)
    return unique_set, len(unique_set)

# Verification
sample = "engineering"
chars, count = extract_unique_chars(sample)
print(f"Input: '{sample}'")
print(f"Unique Characters: {sorted(chars)}")
print(f"Total Unique Count: {count}")
```

## Systems Case Study 1: String Immutability & Buffer Concatenation Complexity

```python
import time

def benchmark_concat(n: int):
    # Naive repeated '+'
    t0 = time.perf_counter()
    s_naive = ""
    for _ in range(n):
        s_naive += "a"
    t_naive = time.perf_counter() - t0
    
    # Optimized str.join()
    t0 = time.perf_counter()
    parts = ["a"] * n
    s_join = "".join(parts)
    t_join = time.perf_counter() - t0
    
    return t_naive, t_join

print(f"{'N':<8} | {'Naive += (sec)':<16} | {'str.join() (sec)':<18} | {'Speedup':<10}")
print("-" * 60)
for n in [5000, 10000, 20000, 40000]:
    tn, tj = benchmark_concat(n)
    speedup = tn / max(tj, 1e-9)
    print(f"{n:<8} | {tn:<16.6f} | {tj:<18.6f} | {speedup:<10.2f}x")
```

## Systems Case Study 2: Dynamic Array Growth Geometry

```python
import sys

def trace_list_growth(max_n: int = 60):
    lst = []
    prev_bytes = sys.getsizeof(lst)
    print(f"{'Appends':<8} | {'Bytes':<8} | {'Slots (Capacity)':<18} | {'Waste Factor':<12}")
    print("-" * 52)
    for i in range(max_n):
        lst.append(i)
        curr_bytes = sys.getsizeof(lst)
        if curr_bytes != prev_bytes or i == 0:
            slots = (curr_bytes - 56) // 8
            waste = slots - (i + 1)
            print(f"{i+1:<8} | {curr_bytes:<8} | {slots:<18} | {waste:<12}")
            prev_bytes = curr_bytes

trace_list_growth(50)
```

## Systems Case Study 3: CPython Hash Table Architecture & Perturbation Probing

```python
def trace_cpython_probing(hash_val: int, table_size: int = 8, steps: int = 8):
    mask = table_size - 1
    perturb = hash_val
    j = hash_val & mask
    trajectory = [j]
    for _ in range(steps - 1):
        j = (5 * j + 1 + perturb) & mask
        perturb >>= 5
        trajectory.append(j)
    return trajectory

h = hash("data_science_2026")
traj = trace_cpython_probing(h, table_size=8, steps=8)
print(f"Hash value: {h}")
print(f"Probing sequence (table size 8): {traj}")
print(f"Unique slots visited: {len(set(traj))} / 8")
```
