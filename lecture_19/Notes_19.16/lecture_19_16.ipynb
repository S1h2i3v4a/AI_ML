**Topics Covered:**

- Chart selection decision matrix
- Data-to-ink ratio and color accessibility
- End-to-end worked solutions to the Day-19 Assignment problems

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Industry standard: Seaborn plotted onto explicit Matplotlib Axes
fig, ax = plt.subplots(figsize=(9, 5))

sns.lineplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    marker='o',
    markersize=8,
    errorbar=None,
    ax=ax
)

ax.set_title('Average Total Bill by Day and Sex (Professional Standard)', fontsize=14, fontweight='bold')
ax.set_xlabel('Day of Week', fontsize=12)
ax.set_ylabel('Total Bill ($)', fontsize=12)
ax.legend(title='Sex')
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Assignment Solution: Tip Pivot Heatmap (Day vs Time)

```python
# Assignment Question: Average tip across day and time
tip_pivot = tips.pivot_table(index='day', columns='time', values='tip', aggfunc='mean')

plt.figure(figsize=(7, 5))
sns.heatmap(tip_pivot, annot=True, fmt='.2f', cmap='coolwarm', cbar_kws={'label': 'Average Tip ($)'})
plt.title('Average Tip: Day vs Time Combination', fontsize=14, fontweight='bold')
plt.xlabel('Dining Time', fontsize=12)
plt.ylabel('Day of Week', fontsize=12)
plt.tight_layout()
plt.show()
```
