# Is the Model Underfitting or Overfitting? (Learning Curves & Diagnostics)

Interactive notebook for Module 24.07: Is the Model Underfitting or Overfitting? (Learning Curves & Diagnostics).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve
from sklearn.linear_model import Ridge
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

# Generate dataset
np.random.seed(42)
m = 100
X = np.sort(np.random.uniform(0, 1, m))
y = np.cos(1.5 * np.pi * X) + np.random.normal(0, 0.2, m)

# Evaluate learning curve
model = make_pipeline(PolynomialFeatures(degree=3), Ridge(alpha=0.1))

train_sizes, train_scores, val_scores = learning_curve(
    model, X[:, np.newaxis], y, cv=5,
    scoring="neg_mean_squared_error",
    train_sizes=np.linspace(0.1, 1.0, 10)
)

train_mse = -np.mean(train_scores, axis=1)
val_mse = -np.mean(val_scores, axis=1)

print("Training Sizes:", train_sizes)
print("Training MSE:  ", train_mse.round(4))
print("Validation MSE:", val_mse.round(4))

```
