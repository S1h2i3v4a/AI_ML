# Lecture 18.10: Add Bar Labels
**Topics Covered:**
- Adding text labels to bars using `plt.text()`
- Positioning coordinates (`x`, `y + offset`, `ha='center'`)
- Preventing label clipping with `plt.ylim()`

```python
import matplotlib.pyplot as plt

years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]  # in $M

plt.figure(figsize=(8, 5))
plt.bar(years, revenue, color='purple', edgecolor='black', width=0.55)
plt.xlabel('Years', fontsize=12)
plt.ylabel('Revenue (in $M)', fontsize=12)
plt.title('Yearly Revenue of Movies with Labels', fontsize=14, fontweight='bold')

# Add labels on top of bars
for i in range(len(revenue)):
    plt.text(years[i], revenue[i] + 20, f'${revenue[i]}M', ha='center', fontweight='bold')

# Add vertical padding to avoid clipping
plt.ylim(0, max(revenue) + 200)
plt.grid(axis='y', linestyle=':', alpha=0.5)
plt.show()
```
