# Lecture 18.16: Multiple Datasets on Scatter Plots
**Topics Covered:**
- Plotting multiple groups on the same scatter canvas
- Using contrasting colors and marker symbols
- Identifying cluster separation (Winter vs Summer weather patterns)

```python
import matplotlib.pyplot as plt

cities = ['City A', 'City B', 'City C', 'City D', 'City E']

# Winter
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.figure(figsize=(8, 5))
plt.scatter(winter_temp, winter_humidity, color='blue', s=80, label='Winter', marker='o')
plt.scatter(summer_temp, summer_humidity, color='orange', s=80, label='Summer', marker='s')

plt.xlabel('Temperature (°C)', fontsize=12)
plt.ylabel('Humidity (%)', fontsize=12)
plt.title('Winter vs Summer: Temperature vs Humidity', fontsize=14, fontweight='bold')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
