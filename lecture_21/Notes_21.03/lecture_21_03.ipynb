# Day 21 - Lecture 21.3: Distance Between Two Points

**Topic:** Distance Between Two Points

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

p = np.array([2.0, 3.0])
q = np.array([6.0, 6.0])

dist_l2 = np.linalg.norm(p - q, 2)
dist_l1 = np.linalg.norm(p - q, 1)

print(f"Euclidean L2: {dist_l2:.4f}")
print(f"Manhattan L1: {dist_l1:.4f}")
```
