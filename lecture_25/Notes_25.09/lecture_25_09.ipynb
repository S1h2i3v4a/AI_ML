# Notes_25.09: kNN - K Nearest Neighbours (Intuition & Distance Metrics)\nNon-parametric lazy learner, Euclidean, Manhattan, and Minkowski distance metrics, choosing K.

```python
import numpy as np

p1 = np.array([1.0, 2.0])
p2 = np.array([4.0, 6.0])

l2_dist = np.linalg.norm(p1 - p2)  # Euclidean
l1_dist = np.sum(np.abs(p1 - p2))  # Manhattan

print(f"Euclidean Distance (L2): {l2_dist:.4f}")
print(f"Manhattan Distance (L1): {l1_dist:.4f}")
```
