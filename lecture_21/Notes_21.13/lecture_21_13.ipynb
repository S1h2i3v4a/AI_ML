# Day 21 - Lecture 21.13: Operations on Matrices

**Topic:** Operations on Matrices

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A @ B:
", A @ B)
print("Inverse of A:
", np.linalg.inv(A))
```
