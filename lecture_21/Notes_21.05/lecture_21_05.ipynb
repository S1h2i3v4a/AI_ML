# Day 21 - Lecture 21.5: Distance Between Parallel Lines & Point-to-Line Distance

**Topic:** Distance Between Parallel Lines & Point-to-Line Distance

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

a, b, c = 3.0, 4.0, -6.0
x0, y0 = 2.0, 3.0
d_pt = abs(a*x0 + b*y0 + c) / np.sqrt(a**2 + b**2)

c1, c2 = -6.0, 9.0
d_par = abs(c1 - c2) / np.sqrt(a**2 + b**2)

print(f"Point-to-line: {d_pt:.3f}")
print(f"Parallel lines: {d_par:.3f}")
```
