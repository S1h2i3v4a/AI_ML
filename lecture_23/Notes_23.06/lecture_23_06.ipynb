# Introduction to Scikit-Learn (sklearn API Design)

Interactive code implementation and visualization.


```python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2.2, 4.1, 5.9, 8.1, 9.8])

model = LinearRegression()
model.fit(X, y)

print(f"Weight (Slope):     {model.coef_[0]:.4f}")
print(f"Bias (Intercept):   {model.intercept_:.4f}")
print("Predicted values:   ", np.round(model.predict(X), 2))
```
