# Day 20 - Lecture 20.14: Random Variables

**Topic:** Random Variables

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

x_vals = [0, 1, 2, 3]
pmf = [0.125, 0.375, 0.375, 0.125]
cdf = np.cumsum(pmf)

print("Discrete RV PMF & CDF:")
for x, p, c in zip(x_vals, pmf, cdf):
    print(f"  x={x} -> p(x)={p:.3f}, F(x)={c:.3f}")
```
