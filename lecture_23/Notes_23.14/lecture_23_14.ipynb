# Evaluation Metrics for Regression (MAE, MSE, RMSE, R-squared & Adjusted R-squared)

Interactive notebook for Module 23.14: Evaluation Metrics for Regression (MAE, MSE, RMSE, R-squared & Adjusted R-squared).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Generate synthetic actual vs predicted data
np.random.seed(42)
m = 100
y_actual = np.random.uniform(20, 100, size=m)
noise = np.random.normal(0, 5, size=m)
y_pred = y_actual + noise

# 1. Compute Metrics
mae = mean_absolute_error(y_actual, y_pred)
mse = mean_squared_error(y_actual, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_actual, y_pred)

# 2. Adjusted R^2 for p = 3 features
p = 3
r2_adj = 1 - ((1 - r2) * (m - 1) / (m - p - 1))

print("Evaluation Metrics Summary:")
print(f"MAE:          {mae:.4f}")
print(f"MSE:          {mse:.4f}")
print(f"RMSE:         {rmse:.4f}")
print(f"R^2 Score:    {r2:.4f}")
print(f"Adjusted R^2: {r2_adj:.4f}")

# Plot Residual Distribution
residuals = y_actual - y_pred

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scatter: Actual vs Predicted
axes[0].scatter(y_actual, y_pred, color="#2563eb", alpha=0.7, edgecolors="k")
axes[0].plot([20, 100], [20, 100], "--r", linewidth=2, label="y = y_hat")
axes[0].set_title(f"Actual vs Predicted (R^2 = {r2:.3f})", fontweight="bold")
axes[0].set_xlabel("Actual Values")
axes[0].set_ylabel("Predicted Values")
axes[0].legend()
axes[0].grid(True, linestyle=":", alpha=0.6)

# Histogram of Residuals
axes[1].hist(residuals, bins=15, color="#10b981", edgecolor="black", alpha=0.8)
axes[1].axvline(0, color="red", linestyle="--", linewidth=2)
axes[1].set_title(f"Residual Distribution (RMSE = {rmse:.3f})", fontweight="bold")
axes[1].set_xlabel("Residual (y - y_hat)")
axes[1].set_ylabel("Frequency")
axes[1].grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```
