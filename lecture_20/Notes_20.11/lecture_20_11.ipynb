# Day 20 - Lecture 20.11: The Law of Total Probability

**Topic:** The Law of Total Probability

This interactive notebook contains reference code and computational experiments for this lecture.

```python
import numpy as np

p_b = np.array([0.50, 0.30, 0.20])
p_a_given_b = np.array([0.01, 0.02, 0.05])
p_total = np.dot(p_b, p_a_given_b)

print(f"Law of Total Probability P(A) = sum(P(A|Bi)P(Bi)) = {p_total:.4f} ({p_total*100:.1f}%)")
```
