# Lecture 18.5: Multiple Datasets on Line Plot
**Topics Covered:**
- Plotting multiple series on a single axes
- Adding `label` parameters
- Configuring `plt.legend()` with custom placement (`loc`)
- Comparing Oscar vs Non-Oscar movie revenues

```python
import matplotlib.pyplot as plt

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.figure(figsize=(9, 5))
plt.plot(years, oscar_revenue, label='Oscar Winners', marker='o', color='blue')
plt.plot(years, non_oscar_revenue, label='Non-Oscar Winners', marker='>', color='green')
plt.title('Years vs Revenue (in $M)', fontsize=14, fontweight='bold')
plt.xlabel('Years', fontsize=12)
plt.ylabel('Revenue (in $M)', fontsize=12)
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
