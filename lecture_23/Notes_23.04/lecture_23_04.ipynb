# Supervised Machine Learning Workflow & Components

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate Underfitting vs Optimal vs Overfitting
np.random.seed(42)
x = np.linspace(-3, 3, 30)
y = x**2 + np.random.randn(30) * 1.2

p_linear = np.poly1d(np.polyfit(x, y, 1))
p_quad = np.poly1d(np.polyfit(x, y, 2))
p_over = np.poly1d(np.polyfit(x, y, 12))

x_dense = np.linspace(-3, 3, 200)

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 4))
ax1.scatter(x, y, color='blue', s=40)
ax1.plot(x_dense, p_linear(x_dense), 'r-', lw=2)
ax1.set_title('Underfitting (High Bias: Degree 1)')
ax1.grid(True, linestyle=':')

ax2.scatter(x, y, color='blue', s=40)
ax2.plot(x_dense, p_quad(x_dense), 'g-', lw=2)
ax2.set_title('Balanced Fit (Optimal: Degree 2)')
ax2.grid(True, linestyle=':')

ax3.scatter(x, y, color='blue', s=40)
ax3.plot(x_dense, p_over(x_dense), 'm-', lw=2)
ax3.set_title('Overfitting (High Variance: Degree 12)')
ax3.set_ylim(-2, 12)
ax3.grid(True, linestyle=':')

plt.tight_layout()
plt.show()
```
