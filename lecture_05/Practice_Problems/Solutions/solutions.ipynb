## Problem 01: Bank Account Management System

```python
class BankAccount:
    """Production-ready bank account model with validation and transaction safeguards."""
    def __init__(self, account_number: str, initial_balance: float = 0.0):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        self.account_number = account_number
        self._balance = float(initial_balance)

    def deposit(self, amount: float) -> bool:
        if amount <= 0:
            print(f"Deposit rejected: Amount INR {amount:,.2f} must be positive.")
            return False
        self._balance += amount
        print(f"Deposited INR {amount:,.2f} | New Balance: INR {self._balance:,.2f}")
        return True

    def withdraw(self, amount: float) -> bool:
        if amount <= 0:
            print(f"Withdrawal rejected: Amount INR {amount:,.2f} must be positive.")
            return False
        if amount > self._balance:
            print(f"Transaction Declined: Insufficient balance (INR {self._balance:,.2f}) for withdrawal of INR {amount:,.2f}.")
            return False
        self._balance -= amount
        print(f"Withdrawn INR {amount:,.2f} | Remaining Balance: INR {self._balance:,.2f}")
        return True

    def check_balance(self) -> float:
        return self._balance

acc = BankAccount("ACC-90210", 1000.0)
print(f"Account: {acc.account_number} | Initial: INR {acc.check_balance():,.2f}")
acc.deposit(500.0)
acc.withdraw(300.0)
acc.withdraw(2000.0)
print(f"Final Balance: INR {acc.check_balance():,.2f}")
```

## Problem 02: Dynamic Review Tracker for Books

```python
from typing import List

class Book:
    """Book entity maintaining title, author, and encapsulated review collection."""
    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author
        self._reviews: List[str] = []

    def add_review(self, review: str) -> None:
        if not review or not isinstance(review, str):
            raise ValueError("Review must be a non-empty string.")
        self._reviews.append(review.strip())
        print(f"Added review #{len(self._reviews)} for '{self.title}'")

    def count_reviews(self) -> int:
        return len(self._reviews)

    def display_reviews(self) -> None:
        print(f"\n--- Reviews for '{self.title}' by {self.author} (Total: {len(self._reviews)}) ---")
        if not self._reviews:
            print("  No reviews submitted yet.")
            return
        for idx, rev in enumerate(self._reviews, start=1):
            print(f"  [{idx}] {rev}")

b = Book("Fluent Python", "Luciano Ramalho")
b.add_review("Essential reading for intermediate to senior engineers.")
b.add_review("Masterclass on the Python Data Model and protocols.")
b.add_review("Crystal clear explanation of decorators and descriptors.")
print(f"Total count: {b.count_reviews()}")
b.display_reviews()
```

## Problem 03: Encapsulated Student Academic Record

```python
class Student:
    """Encapsulated Student entity with validated properties."""
    def __init__(self, name: str, roll_no: str, marks: float):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    @property
    def name(self) -> str:
        return self._name

    @name.setter
    def name(self, val: str) -> None:
        if not val or not isinstance(val, str):
            raise ValueError("Student name must be a valid non-empty string.")
        self._name = val.strip()

    @property
    def roll_no(self) -> str:
        return self._roll_no

    @roll_no.setter
    def roll_no(self, val: str) -> None:
        if not val or not isinstance(val, str):
            raise ValueError("Roll number must be a valid non-empty string.")
        self._roll_no = val.strip()

    @property
    def marks(self) -> float:
        return self._marks

    @marks.setter
    def marks(self, val: float) -> None:
        if not isinstance(val, (int, float)) or not (0.0 <= val <= 100.0):
            raise ValueError(f"Marks must be in [0.0, 100.0], received: {val}")
        self._marks = float(val)

    def calculate_grade(self) -> str:
        m = self._marks
        if m >= 90.0:
            return "A+"
        elif m >= 80.0:
            return "A"
        elif m >= 70.0:
            return "B"
        elif m >= 60.0:
            return "C"
        elif m >= 50.0:
            return "D"
        else:
            return "F"

s = Student("Vikramaditya", "CS2026-088", 87.5)
print(f"Student: {s.name} | Roll: {s.roll_no} | Marks: {s.marks} | Grade: {s.calculate_grade()}")
s.marks = 94.0
print(f"Updated Marks: {s.marks} | Updated Grade: {s.calculate_grade()}")
try:
    s.marks = 105.0
except ValueError as e:
    print("Handled validation error:", e)
```

## Problem 04: Hierarchical Shape Geometry & Area Polymorphism

```python
import math
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self) -> float:
        pass

class Circle(Shape):
    def __init__(self, radius: float):
        if radius <= 0:
            raise ValueError("Radius must be strictly positive.")
        self.radius = radius

    def area(self) -> float:
        return math.pi * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        if width <= 0 or height <= 0:
            raise ValueError("Dimensions must be strictly positive.")
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height

class Triangle(Shape):
    def __init__(self, base: float, height: float):
        if base <= 0 or height <= 0:
            raise ValueError("Dimensions must be strictly positive.")
        self.base = base
        self.height = height

    def area(self) -> float:
        return 0.5 * self.base * self.height

shapes = [Circle(7.0), Rectangle(10.0, 5.0), Triangle(8.0, 6.0)]
for shape in shapes:
    print(f"Class: {shape.__class__.__name__:<12} | Computed Area: {shape.area():.4f}")
```

## Problem 05: Fleet Vehicle Taxonomy & Property Extension

```python
class Vehicle:
    def __init__(self, brand: str, model: str):
        self.brand = brand
        self.model = model

    def get_details(self) -> str:
        return f"Vehicle: {self.brand} {self.model}"

class Car(Vehicle):
    def __init__(self, brand: str, model: str, seats: int):
        super().__init__(brand, model)
        if seats <= 0:
            raise ValueError("Seats must be positive.")
        self.seats = seats

    def get_details(self) -> str:
        parent_details = super().get_details()
        return f"{parent_details} | Type: Passenger Car | Seating Capacity: {self.seats}"

class Bike(Vehicle):
    def __init__(self, brand: str, model: str, engine_cc: int):
        super().__init__(brand, model)
        if engine_cc <= 0:
            raise ValueError("Displacement must be positive.")
        self.engine_cc = engine_cc

    def get_details(self) -> str:
        parent_details = super().get_details()
        return f"{parent_details} | Type: Motorcycle | Engine Displacement: {self.engine_cc}cc"

c = Car("Tesla", "Model S Plaid", seats=5)
b = Bike("Ducati", "Panigale V4", engine_cc=1103)
print(c.get_details())
print(b.get_details())
```

## Problem 06: Abstract Enterprise Payroll Engine

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, emp_id: str, name: str):
        self.emp_id = emp_id
        self.name = name

    @abstractmethod
    def calculate_salary(self) -> float:
        pass

class Intern(Employee):
    def __init__(self, emp_id: str, name: str, stipend: float):
        super().__init__(emp_id, name)
        self.stipend = stipend

    def calculate_salary(self) -> float:
        return float(self.stipend)

class FullTimeEmployee(Employee):
    def __init__(self, emp_id: str, name: str, base: float, hra: float, allowance: float):
        super().__init__(emp_id, name)
        self.base = base
        self.hra = hra
        self.allowance = allowance

    def calculate_salary(self) -> float:
        return float(self.base + self.hra + self.allowance)

class ContractEmployee(Employee):
    def __init__(self, emp_id: str, name: str, hourly_rate: float, hours_worked: float):
        super().__init__(emp_id, name)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self) -> float:
        return float(self.hourly_rate * self.hours_worked)

payroll = [
    Intern("INT-101", "Aarav Sharma", stipend=25000.0),
    FullTimeEmployee("FTE-202", "Diya Nair", base=85000.0, hra=20000.0, allowance=15000.0),
    ContractEmployee("CON-303", "Kabir Mehta", hourly_rate=1200.0, hours_worked=160.0)
]

print(f"{'ID':<10} | {'Name':<16} | {'Role':<20} | {'Calculated Net Salary':<20}")
print("-" * 72)
for emp in payroll:
    print(f"{emp.emp_id:<10} | {emp.name:<16} | {emp.__class__.__name__:<20} | INR {emp.calculate_salary():,.2f}")
```

## Problem 07: Pythonic Constructor Overloading Simulation

```python
from typing import Optional
from datetime import datetime

class Person:
    def __init__(self, name: str, age: Optional[int] = None, address: Optional[str] = None):
        self.name = name
        self.age = age
        self.address = address or "Not Specified"

    @classmethod
    def from_birth_year(cls, name: str, birth_year: int, address: Optional[str] = None):
        current_year = datetime.now().year
        age = current_year - birth_year
        return cls(name, age=age, address=address)

    def __repr__(self) -> str:
        return f"Person(name='{self.name}', age={self.age}, address='{self.address}')"

p1 = Person("Aanya")
p2 = Person("Rohan", age=28)
p3 = Person("Neha", age=32, address="Bengaluru, KA")
p4 = Person.from_birth_year("Karthik", 1998, address="Hyderabad, TS")

print(p1)
print(p2)
print(p3)
print(p4)
```

## Problem 08: Real-Time Fleet Census via Class Attributes

```python
class Player:
    player_count: int = 0

    def __init__(self, username: str, rank: str = "Bronze"):
        self.username = username
        self.rank = rank
        Player.player_count += 1
        self._is_active = True
        print(f"Player '{self.username}' connected (Total: {Player.player_count})")

    @classmethod
    def get_active_count(cls) -> int:
        return cls.player_count

    def terminate_session(self) -> None:
        if self._is_active:
            Player.player_count -= 1
            self._is_active = False
            print(f"Player '{self.username}' disconnected (Total: {Player.player_count})")

print("Initial active players:", Player.get_active_count())
p1 = Player("ShadowNinja", "Diamond")
p2 = Player("CyberValkyrie", "Master")
p3 = Player("PixelMage", "Gold")
print("Active count mid-game:", Player.get_active_count())
p2.terminate_session()
print("Active count after logout:", Player.get_active_count())
```

## Problem 09: Omnivore Diamond Inheritance & C3 Linearization

```python
class Organism:
    def __init__(self, kingdom: str = "Animalia", **kwargs):
        super().__init__(**kwargs)
        self.kingdom = kingdom

    def feed(self) -> str:
        return f"[Kingdom: {self.kingdom}]"

class Herbivore(Organism):
    def __init__(self, plant_diet: str = "berries and roots", **kwargs):
        super().__init__(**kwargs)
        self.plant_diet = plant_diet

    def feed(self) -> str:
        return f"{super().feed()} + Eats plants: {self.plant_diet}"

class Carnivore(Organism):
    def __init__(self, meat_diet: str = "salmon and rodents", **kwargs):
        super().__init__(**kwargs)
        self.meat_diet = meat_diet

    def feed(self) -> str:
        return f"{super().feed()} + Hunts meat: {self.meat_diet}"

class Bear(Herbivore, Carnivore):
    def __init__(self, species_name: str, **kwargs):
        super().__init__(**kwargs)
        self.species_name = species_name

    def feed(self) -> str:
        return f"Bear ({self.species_name}) Omnivore Diet: {super().feed()}"

grizzly = Bear("Grizzly Bear")
print(grizzly.feed())
print("\nMethod Resolution Order (MRO):")
for idx, cls in enumerate(Bear.__mro__, start=1):
    print(f"  {idx}. {cls.__name__}")
```

## Problem 10: Mini-Project: Multi-Class OOP Chat System

```python
from typing import List, Dict, Optional
from datetime import datetime

class Message:
    def __init__(self, sender: str, recipient: Optional[str], content: str):
        self.sender = sender
        self.recipient = recipient
        self.content = content
        self.timestamp = datetime.now()

    def __str__(self) -> str:
        ts = self.timestamp.strftime("%H:%M:%S")
        target = f" -> @{self.recipient}" if self.recipient else " -> #all"
        return f"[{ts}] [{self.sender}{target}]: {self.content}"

class User:
    def __init__(self, username: str):
        self.username = username
        self.inbox: List[Message] = []

    def receive_message(self, message: Message) -> None:
        self.inbox.append(message)

    def print_inbox(self) -> None:
        print(f"\n--- Inbox for @{self.username} ({len(self.inbox)} messages) ---")
        for msg in self.inbox:
            print(f"  {msg}")

class ChatRoom:
    def __init__(self, room_name: str):
        self.room_name = room_name
        self.members: Dict[str, User] = {}
        self.history: List[Message] = []

    def join(self, user: User) -> None:
        self.members[user.username] = user
        print(f"@{user.username} entered #{self.room_name}")

    def broadcast(self, sender: User, content: str) -> None:
        msg = Message(sender.username, None, content)
        self.history.append(msg)
        for member in self.members.values():
            member.receive_message(msg)

    def direct_message(self, sender: User, recipient_name: str, content: str) -> None:
        recipient = self.members.get(recipient_name)
        if not recipient:
            print(f"Error: @{recipient_name} is not in #{self.room_name}")
            return
        msg = Message(sender.username, recipient_name, content)
        self.history.append(msg)
        recipient.receive_message(msg)
        sender.receive_message(msg)

    def print_history(self) -> None:
        print(f"\n=== History for #{self.room_name} ({len(self.history)} messages) ===")
        for msg in self.history:
            print(f"  {msg}")

room = ChatRoom("dev-ops-hq")
u1 = User("Alice")
u2 = User("Bob")
u3 = User("Charlie")

room.join(u1)
room.join(u2)
room.join(u3)

room.broadcast(u1, "Deployment to production starting now.")
room.broadcast(u2, "Monitoring dashboards look green.")
room.direct_message(u1, "Charlie", "Please verify database connection pool limits.")
room.print_history()
u3.print_inbox()
```

## Systems Case Study 1: CPython Object Memory & `__slots__` Anatomy

```python
import sys
import tracemalloc

class StandardPoint:
    def __init__(self, x: float, y: float, z: float):
        self.x = x
        self.y = y
        self.z = z

class SlottedPoint:
    __slots__ = ('x', 'y', 'z')
    def __init__(self, x: float, y: float, z: float):
        self.x = x
        self.y = y
        self.z = z

p_std = StandardPoint(1.0, 2.0, 3.0)
p_slot = SlottedPoint(1.0, 2.0, 3.0)

std_size = sys.getsizeof(p_std) + sys.getsizeof(p_std.__dict__)
slot_size = sys.getsizeof(p_slot)

print(f"StandardPoint instance overhead: {std_size} bytes (Object: {sys.getsizeof(p_std)}B + Dict: {sys.getsizeof(p_std.__dict__)}B)")
print(f"SlottedPoint instance overhead:  {slot_size} bytes")
print(f"Direct instance savings:         {(1 - slot_size / std_size) * 100:.2f}%")

# Benchmark with 100,000 instances
def measure_fleet(cls, n=100_000):
    tracemalloc.start()
    fleet = [cls(float(i), float(i*2), float(i*3)) for i in range(n)]
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    return peak / (1024 * 1024)

mem_std = measure_fleet(StandardPoint)
mem_slot = measure_fleet(SlottedPoint)
print(f"\nHeap allocation for 100k instances:")
print(f"  StandardPoint: {mem_std:.2f} MB")
print(f"  SlottedPoint:  {mem_slot:.2f} MB")
print(f"  Total Heap Reduction: {(1 - mem_slot / mem_std) * 100:.2f}%")
```

## Systems Case Study 2: Diamond Inheritance & C3 Linearization Algorithm

```python
def c3_merge(sequences):
    """Pure Python implementation of C3 Linearization monotonic merge."""
    result = []
    seqs = [list(s) for s in sequences]
    while True:
        # Filter out empty sequences
        seqs = [s for s in seqs if s]
        if not seqs:
            return result
        # Find candidate head not in the tail of any sequence
        for s in seqs:
            candidate = s[0]
            not_in_tails = True
            for other_s in seqs:
                if candidate in other_s[1:]:
                    not_in_tails = False
                    break
            if not_in_tails:
                break
        else:
            raise TypeError("Inconsistent hierarchy: Cannot compute monotonic C3 MRO!")
        
        result.append(candidate)
        for s in seqs:
            if s[0] == candidate:
                s.pop(0)

# Verify against Python's built-in MRO for Bear
class O: pass
class A(O): pass
class B(O): pass
class D(A, B): pass

computed = c3_merge([[D], [A, O, object], [B, O, object], [A, B]])
native = list(D.__mro__)
print("Simulated C3 Merge result: ", [cls.__name__ for cls in computed])
print("Native Python D.__mro__:    ", [cls.__name__ for cls in native])
assert [cls.__name__ for cls in computed] == [cls.__name__ for cls in native]
print("Assertion Passed: Simulated C3 algorithm matches CPython MRO perfectly!")
```

## Systems Case Study 3: Dynamic Attribute Lookup Protocol & Descriptors

```python
class DataDescriptor:
    def __get__(self, instance, owner):
        return "1. Value from Data Descriptor (__get__)"
    def __set__(self, instance, value):
        pass

class NonDataDescriptor:
    def __get__(self, instance, owner):
        return "3. Value from Non-Data Descriptor (__get__)"

class Target:
    data_desc = DataDescriptor()
    non_data_desc = NonDataDescriptor()

    def __init__(self):
        # Override descriptors in instance dictionary
        self.__dict__['data_desc'] = "Overridden in instance __dict__"
        self.__dict__['non_data_desc'] = "2. Overridden in instance __dict__"
        self.__dict__['regular_attr'] = "Value in instance __dict__"

    def __getattr__(self, name):
        return f"4. Value from fallback __getattr__('{name}')"

obj = Target()

print("Lookup data_desc (Data descriptor wins over instance dict):")
print(" ->", obj.data_desc)

print("\nLookup non_data_desc (Instance dict wins over non-data descriptor):")
print(" ->", obj.non_data_desc)

print("\nLookup non_existent_attr (Fallback to __getattr__):")
print(" ->", obj.non_existent_attr)
```
