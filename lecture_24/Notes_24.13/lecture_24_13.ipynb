# Logistic Regression: Intuition & The Sigmoid Function

Interactive notebook for Module 24.13: Logistic Regression: Intuition & The Sigmoid Function.


```python
import numpy as np
import matplotlib.pyplot as plt

# Sigmoid function implementation
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

# Generate z domain
z = np.linspace(-8, 8, 200)
sigma = sigmoid(z)

# Plot Sigmoid Curve
plt.figure(figsize=(8, 5))
plt.plot(z, sigma, color="#2563eb", linewidth=2.5, label=r"$\sigma(z) = \frac{1}{1 + e^{-z}}$")
plt.axhline(0.5, color="#dc2626", linestyle="--", alpha=0.7, label="Threshold p = 0.5 (z = 0)")
plt.axvline(0, color="gray", linestyle=":", alpha=0.5)

plt.title("The Standard Logistic Sigmoid Activation Function", fontsize=13, fontweight="bold")
plt.xlabel("Logit Input (z = w^T x + b)", fontsize=11)
plt.ylabel("Predicted Probability P(y = 1 | x)", fontsize=11)
plt.ylim(-0.05, 1.05)
plt.legend(frameon=True, fontsize=11)
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()

```
