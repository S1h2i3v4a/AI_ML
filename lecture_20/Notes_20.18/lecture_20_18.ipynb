# Day 20 - Lecture 20.18: Binomial Distribution

**Topic:** Binomial Distribution

This interactive notebook contains reference code and computational experiments for this lecture.

```python
from scipy.stats import binom

n = 5
k = 3
p = 0.5

prob = binom.pmf(k, n, p)
print(f"Binomial PMF (k={k}, n={n}, p={p}) = {prob:.4f} (5/16)")
```
