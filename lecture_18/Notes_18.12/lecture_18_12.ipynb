# Lecture 18.12: Horizontal Bar Charts
**Topics Covered:**
- Concept of horizontal bar charts (`plt.barh`)
- Solving the long category label problem
- Mapping values to `width` rather than `height`

```python
import matplotlib.pyplot as plt

oscar_movies = ['The Dark Knight', 'The Hurt Locker', "The King's Speech", 'The Artist', 'Argo']
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.figure(figsize=(9, 5))
plt.barh(oscar_movies, oscar_revenue, color='teal', edgecolor='black', height=0.6)
plt.xlabel('Revenue (in $M)', fontsize=12)
plt.ylabel('Movie', fontsize=12)
plt.title('Revenue for each Movie (Horizontal Bar Chart)', fontsize=14, fontweight='bold')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
