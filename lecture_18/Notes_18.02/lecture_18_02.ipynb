# Lecture 18.2: How to Plot Data
**Topics Covered:**
- Cartesian 2D coordinate system (X and Y coordinates)
- Fundamental `plt.plot(X, Y)` syntax
- Importance of array length matching and ordering

```python
import matplotlib.pyplot as plt

# Simple coordinate pairs
X = [1, 2, 3, 4]
Y = [5, 6, 7, 8]

plt.plot(X, Y)
plt.title('Basic Plot: (1,5) to (4,8)')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.show()
```
