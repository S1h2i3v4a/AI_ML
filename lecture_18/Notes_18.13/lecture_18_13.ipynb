# Lecture 18.13: Scatter Plots
**Topics Covered:**
- Concept and utility of scatter plots
- Bivariate relationship and correlation detection
- Case study: Patient Age vs Systolic Blood Pressure

```python
import matplotlib.pyplot as plt

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.figure(figsize=(8, 5))
plt.scatter(age, blood_pressure, color='crimson', s=70, edgecolor='black')
plt.title('Age vs Blood Pressure', fontsize=14, fontweight='bold')
plt.xlabel('Age (Years)', fontsize=12)
plt.ylabel('Blood Pressure (mmHg)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
