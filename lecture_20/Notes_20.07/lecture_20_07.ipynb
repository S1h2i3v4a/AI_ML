# Day 20 - Lecture 20.7: The Addition Rule of Probability

**Topic:** The Addition Rule of Probability

This interactive notebook contains reference code and computational experiments for this lecture.

```python
p_a = 0.40
p_b = 0.30
p_cap = 0.15

p_union = p_a + p_b - p_cap
p_neither = 1.0 - p_union

print(f"P(A U B) = {p_a} + {p_b} - {p_cap} = {p_union:.2f}")
print(f"P(Neither) = 1 - {p_union:.2f} = {p_neither:.2f}")
```
