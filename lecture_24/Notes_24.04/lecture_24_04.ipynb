# Overfitting (High Variance & Memorizing Noise)

Interactive notebook for Module 24.04: Overfitting (High Variance & Memorizing Noise).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Generate synthetic non-linear ground truth with noise
np.random.seed(42)
m = 15
X = np.sort(np.random.uniform(0, 1, m))
y = np.cos(1.5 * np.pi * X) + np.random.normal(0, 0.15, m)

# Test set
X_test = np.linspace(0, 1, 100)
y_test = np.cos(1.5 * np.pi * X_test)

# High-degree polynomial (Degree 14 on 15 samples = Overfitting)
overfit_model = make_pipeline(PolynomialFeatures(degree=14), LinearRegression())
overfit_model.fit(X[:, np.newaxis], y)

y_pred_train = overfit_model.predict(X[:, np.newaxis])
y_pred_test = overfit_model.predict(X_test[:, np.newaxis])

train_mse = mean_squared_error(y, y_pred_train)
test_mse = mean_squared_error(y_test, y_pred_test)

print(f"Training MSE (Near Zero): {train_mse:.5f}")
print(f"Test MSE (Exploded):      {test_mse:.5f}")

```
