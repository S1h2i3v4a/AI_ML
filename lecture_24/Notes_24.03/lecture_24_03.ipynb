# Other Feature Engineering Techniques (Scaling, Binning & Interactions)

Interactive notebook for Module 24.03: Other Feature Engineering Techniques (Scaling, Binning & Interactions).


```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler, PolynomialFeatures

# Sample feature data
np.random.seed(42)
df = pd.DataFrame({
    "Age": np.random.randint(20, 65, size=5),
    "Salary": np.array([25000, 45000, 120000, 85000, 310000]),
    "CreditScore": np.array([620, 710, 790, 640, 820])
})

print("Original Features:")
print(df)

# 1. StandardScaler
scaler = StandardScaler()
df_standardized = pd.DataFrame(scaler.fit_transform(df), columns=[f"{c}_std" for c in df.columns])
print("\nStandardized Features (Mean ~ 0, Std = 1):")
print(df_standardized.round(3))

# 2. Polynomial & Interaction Terms
poly = PolynomialFeatures(degree=2, include_bias=False)
poly_features = poly.fit_transform(df[["Age", "Salary"]])
poly_df = pd.DataFrame(poly_features, columns=poly.get_feature_names_out(["Age", "Salary"]))
print("\nPolynomial & Interaction Features (Degree 2):")
print(poly_df.head(2))

```
