# Operations on Functions: Scalar Multiplication & Addition (Vertical Transformations)

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 300)
y_base = x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y_base, 'k-', lw=2, label='f(x) = x^2')
plt.plot(x, 2 * y_base, 'r--', lw=2, label='2 * f(x) (Stretched)')
plt.plot(x, y_base + 3, 'b-.', lw=2, label='f(x) + 3 (Shifted Up)')
plt.title('Vertical Transformations on f(x) = x^2')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
