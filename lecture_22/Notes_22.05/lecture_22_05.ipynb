# Input Transformations: Scaling & Shifts (Horizontal Transformations & Affine Maps)

Interactive code implementation and visualization.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 6, 300)
y_base = x**2

plt.figure(figsize=(8, 5))
plt.plot(x, y_base, 'k-', lw=2, label='f(x) = x^2')
plt.plot(x, (x - 2)**2, 'r--', lw=2, label='f(x - 2) (Shift Right)')
plt.plot(x, (x + 2)**2, 'b-.', lw=2, label='f(x + 2) (Shift Left)')
plt.title('Horizontal Input Shifts on f(x) = x^2')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```
