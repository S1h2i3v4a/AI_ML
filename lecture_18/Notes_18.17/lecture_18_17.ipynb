# Lecture 18.17: Pie Charts
**Topics Covered:**
- Concept of Pie Charts (part-to-whole proportion)
- `plt.pie()` syntax and parameter overview
- Displaying percentages automatically with `autopct='%1.1f%%'`
- Assigning custom palettes

```python
import matplotlib.pyplot as plt

expenses = ['Salaries', 'Rent', 'Marketing', 'R&D', 'Miscellaneous']
amounts = [500, 150, 120, 100, 50]
colors = ['purple', 'green', 'blue', 'pink', 'yellow']

plt.figure(figsize=(7, 7))
plt.pie(amounts, labels=expenses, autopct='%1.1f%%', colors=colors)
plt.title('Corporate Expense Allocation', fontsize=14, fontweight='bold')
plt.show()
```
