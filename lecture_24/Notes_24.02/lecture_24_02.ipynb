# The Dummy Variable Trap & Multicollinearity

Interactive notebook for Module 24.02: The Dummy Variable Trap & Multicollinearity.


```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Demonstration of Dummy Variable Trap
df = pd.DataFrame({
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female"],
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [30, 45, 55, 65, 80, 95]
})

print("1. Trapped Encoding (Both Dummies Kept):")
df_trapped = pd.get_dummies(df, columns=["Gender"], drop_first=False)
print(df_trapped)

# Calculate matrix condition number to verify multicollinearity
X_trapped = np.c_[np.ones(len(df)), df_trapped[["Experience", "Gender_Female", "Gender_Male"]].values.astype(float)]
cond_trapped = np.linalg.cond(X_trapped.T @ X_trapped)
print(f"Condition Number (Trapped - Highly Ill-conditioned): {cond_trapped:.2e}")

print("\n2. Proper Encoding (drop_first=True):")
df_proper = pd.get_dummies(df, columns=["Gender"], drop_first=True)
print(df_proper)

X_proper = np.c_[np.ones(len(df)), df_proper[["Experience", "Gender_Male"]].values.astype(float)]
cond_proper = np.linalg.cond(X_proper.T @ X_proper)
print(f"Condition Number (Proper - Well-conditioned): {cond_proper:.2f}")

# Fit model on proper data
model = LinearRegression().fit(df_proper[["Experience", "Gender_Male"]], df_proper["Salary"])
print(f"\nModel Intercept (Female Baseline): {model.intercept_:.2f}")
print(f"Coefficients: Experience={model.coef_[0]:.2f}, Male_Delta={model.coef_[1]:.2f}")

```
