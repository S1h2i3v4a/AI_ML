```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -250, 250)))

def binary_cross_entropy(y_true, y_pred, eps=1e-15):
    # Clip probabilities to prevent log(0) undefined numerical errors
    y_pred = np.clip(y_pred, eps, 1.0 - eps)
    m = len(y_true)
    return - (1.0 / m) * np.sum(y_true * np.log(y_pred) + (1.0 - y_true) * np.log(1.0 - y_pred))

# Test cross entropy loss
y_true = np.array([1, 0, 1, 1, 0])
y_perfect = np.array([0.99, 0.01, 0.98, 0.95, 0.02])
y_terrible = np.array([0.05, 0.92, 0.10, 0.02, 0.88])

print(f"Loss with accurate predictions: {binary_cross_entropy(y_true, y_perfect):.4f}")
print(f"Loss with terrible predictions: {binary_cross_entropy(y_true, y_terrible):.4f}")

```
