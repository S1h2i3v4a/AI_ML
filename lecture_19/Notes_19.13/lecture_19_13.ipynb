# Lecture 19.13: Categorical Plots in Seaborn
**Topics Covered:**
- Categorical vs Continuous data
- Bar plots with mean aggregation (`sns.barplot()`)
- Statistical dispersion with `sns.boxplot()`
- Grouping by secondary category with `hue`

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# 1. Bar Plot (Mean estimation)
plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x='day', y='total_bill', hue='sex', palette='Blues')
plt.title('Average Total Bill by Day and Sex', fontsize=13, fontweight='bold')
plt.show()

# 2. Box Plot (Five-number summary across days)
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Total Bill Distribution by Day', fontsize=13, fontweight='bold')
plt.show()
```
